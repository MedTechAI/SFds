#### _1. Введение в статистический анализ медицинских данных_

---
<span style="font-size:0.8em;"> [TOC](#toc) </span>

Приветствуем вас на дисциплине «Статистический анализ медицинских данных». В рамках первого модуля вы познакомитесь с основными понятиями статистики:

генеральная совокупность,
выборка,
описательная статистика.
Статистический анализ играет ключевую роль в современной медицине, помогая извлекать ценную информацию из огромного объема данных, накопленных в процессе исследований, диагностики и лечения. Статистика позволяет обрабатывать и интерпретировать эти данные, выявлять закономерности, оценивать эффективность различных методов лечения, определять факторы риска и делать прогнозы.

Медицинская статистика — это область статистики, которая специализируется на сборе, анализе и интерпретации данных, связанных с медицинскими исследованиями, заболеваниями, лечением и здоровьем человека. Она помогает врачам и исследователям принимать обоснованные решения на основе фактических данных и статистических закономерностей.

Медицинская статистика включает в себя различные аспекты:

сбор данных,
их классификацию,
анализ,
интерпретацию,
визуализацию.
Статистические методы позволяют определить связи между различными факторами здоровья, оценить эффективность лечения и качество медицинской помощи, спрогнозировать риски заболеваний.

В медицинской статистике широко используются такие понятия, как среднее значение, медиана, мода, стандартное отклонение, корреляция, регрессионный анализ, доверительные интервалы и многие другие. Эти методы позволяют не только описать данные и выявить закономерности, но и проверить статистическую значимость полученных результатов.

Одно из ключевых направлений медицинской статистики ― клинические исследования, в рамках которых анализируются эффективность и безопасность новых методов лечения, препаратов и медицинских технологий. Также статистический анализ медицинских данных используется для мониторинга эпидемиологической ситуации, изучения распространенности заболеваний и оценки показателей здоровья населения.

История
Когда медицина только начинала свое становление как наука и искусство, врачи и ученые стремились понять природу болезней, их причины и способы лечения. Однако без систематического подхода к анализу данных и статистике.

С развитием науки и техники появились первые попытки применения статистических методов в медицине. Одним из первых выдающихся ученых, которые внесли вклад в развитие медицинской статистики, был Джон Сноу. Он исследовал распространение холеры в Лондоне, используя методы анализа статистических данных для выявления источника заболевания. Оказалось, что им была питьевая вода из колодцев на улице Брод стрит.

С появлением медицинских школ и клиник ученые стали активно изучать большие объемы клинических данных, но для их анализа требовались новые инструменты. Именно тогда зародилась идея применения статистических методов, таких как корреляционный анализ, регрессионный анализ и методы многомерного анализа, к медицинским данным.

В XIX веке развитие медицинской статистики получило мощный импульс благодаря выдающемуся ученому и публицисту Уильяму Фарру, который предложил использовать статистику для изучения эпидемиологии и распространения болезней. Фарр провел множество исследований, показавших важность статистического анализа данных для понимания закономерностей в медицине.

В XX веке с развитием компьютерных технологий и статистических программ медицинская статистика стала неотъемлемой частью практики врачей и исследователей. С появлением мощных инструментов анализа данных, таких как SPSS, SAS, R и другие, ученые получили возможность проводить более глубокие и точные исследования в области медицины.

Благодаря постоянному развитию методов статистического анализа, медицинская статистика продолжает менять наш взгляд на здоровье и болезни, делая медицину более эффективной и научно обоснованной.

##### Государственная медицинская статистика
Государственная медицинская статистика представляет собой обширную область статистики, связанную со сбором, обработкой и анализом данных о здоровье населения, медицинской помощи и заболеваниях. В рамках государственной статистики выделяются несколько разделов, каждый из которых играет важную роль в оценке и управлении здравоохранением на уровне страны или региона.

Эпидемиологическая статистика ― занимается анализом распространенности и распределения различных заболеваний среди населения, выявлением факторов риска и мер по их контролю. Эпидемиологическая статистика дает информацию о заболеваемости, смертности, вакцинации и других аспектах общественного здоровья.

Статистика медицинских учреждений ― включает в себя информацию об активности больниц, поликлиник, амбулаторий, а также о качестве и объеме медицинской помощи, предоставляемой населению. Данные этого раздела помогают оценить доступность медицинских услуг, нагрузку на здравоохранение, эффективность работы медицинских учреждений.

Статистика здоровья населения ― включает в себя информацию о физиологических показателях здоровья, образе жизни, привычках, смертности и продолжительности жизни. Эти данные позволяют проводить мониторинг здоровья населения, выявлять тенденции и проблемы в здравоохранении.

Статистика медицинских работников ― этот раздел содержит информацию о квалификации, численности, распределении и оплате труда медицинского персонала. Данные о медицинских работниках необходимы для планирования кадрового состава в здравоохранении и обеспечения его эффективной работы.

Статистика медицинского образования и научных исследований ― этот раздел включает данные о численности выпускников медицинских учебных заведений, направлениях исследований, научных публикациях и инновациях в медицине. Эти данные помогают понять состояние медицинской науки и образования, определить приоритеты развития и совершенствования медицинской отрасли.

Все эти разделы государственной медицинской статистики работают взаимосвязанно, обеспечивая комплексное представление о состоянии здравоохранения и населения, а также позволяя принимать обоснованные решения на уровне государства или региона.

##### Виды медицинской статистики
В рамках медицинской статистики выделяют несколько основных видов, каждый из которых имеет свои особенности и применение.

1. Дескриптивная (описательная) статистика используется для описания и суммирования данных без вывода общих заключений. Дескриптивная статистика позволяет представить информацию в виде числовых показателей, графиков или диаграмм, что облегчает восприятие и анализ данных в медицинских исследованиях.

2. Инференциальная статистика применяется для вывода общих заключений и обобщений на основе анализа выборочных данных. Инференциальная статистика играет важную роль в проведении клинических исследований и определении эффективности лечения, поскольку позволяет делать обобщения о целой популяции на основе данных, собранных из выборки.

3. Биостатистика специализируется на анализе данных, связанных с биологическими и медицинскими исследованиями. Биостатистика включает в себя методы анализа клинических испытаний, эпидемиологических исследований, генетических данных и других областей медицины, где требуется статистическая обработка информации.

4. Эпидемиологическая статистика направлена на изучение распространения болезней в популяции, выявление факторов риска и оценку воздействия общественного здравоохранения. Эпидемиологическая статистика используется для анализа эпидемиологических данных, создания моделей распространения заболеваний и разработки мероприятий по их контролю и профилактике.

Итоги
Статистика помогает врачам оценивать вероятности возникновения различных заболеваний, определять эффективность лечения, оценивать риски для пациентов и принимать взвешенные решения на основе объективных данных.

Знание методов статистического анализа позволяет проводить исследования с высокой степенью достоверности, учитывая различные факторы влияния и контролируя ошибки. Анализ корреляции, регрессии, дисперсии и другие методы помогают выявлять взаимосвязи и закономерности, оптимизировать лечебные подходы и улучшать качество медицинской помощи.

Медицинская статистика играет важную роль в интерпретации результатов лабораторных тестов и клинических наблюдений, помогая врачам и исследователям делать выводы на основе объективных данных. Статистические методы позволяют определить, насколько вероятны полученные результаты при условии, что нулевая гипотеза (например, отсутствие различий между группами) верна. Это помогает решить, действительно ли обнаруженные различия или ассоциации являются статистически значимыми.

Используя доверительные интервалы для оценки неопределенности вокруг средних значений или оценок параметров, медицинская статистика помогает определить точность измерений и дает представление о возможном диапазоне, в котором могут находиться истинные значения. С помощью медицинской статистики можно оценить чувствительность и специфичность диагностических тестов, что важно для оценки их точности и применимости в клинической практике.

Таким образом, знание статистики и эпидемиологии позволяет:

анализировать данные, оценивать риски и преимущества различных методов лечения;
принимать обоснованные решения в диагностике и лечении пациентов;
определять факторы риска для заболеваний и разрабатывать эффективные стратегии профилактики, что способствует снижению заболеваемости и смертности;
оценивать потребности в здравоохранении, оптимизировать ресурсы и разрабатывать эффективные программы по улучшению здоровья населения;
мониторить качество медицинской помощи, выявлять проблемные области и вносить коррективы для улучшения качества услуг.

#### _2. Генеральная совокупность, выборка_

---
<span style="font-size:0.8em;"> [TOC](#toc) </span>

Объектом исследования в статистике является статистическая совокупность ― группа, состоящая из большого числа относительно похожих единиц (больных каким-либо заболеванием, случаев неудовлетворенности медицинской помощью, пациентов, получивших инвалидность, случаев впервые возникшего заболевания).

Сначала исследователь изучает каждую единицу совокупности, выбирая те признаки, которые, по его мнению, помогут наилучшим образом осветить изучаемую проблему. Затем полученные сведения обобщаются и анализируются в целом по всей совокупности. Для этого используются методы статистики.

##### Генеральная совокупность ― все изучаемые объекты того вида, над которым исследователи проводят наблюдения в целях получить конкретные значения случайной величины. Например, если мы тестируем новое лекарство от некоторого заболевания, то генеральной совокупностью для нас будут абсолютно все пациенты, страдающие данным заболеванием.

Часто на практике ученые работают только с небольшим количеством случайно отобранных объектов из изучаемой генеральной совокупности, называемых выборкой (выборочной совокупностью).

Пример
Цель исследования ― изучить здоровье студентов Сеченовского университета.

Генеральная совокупность ― все студенты Сеченовского университета на данный момент. В 2024 году объем совокупности составляет примерно восемнадцать тысяч человек (N=18 000).

Выборочная совокупность ― двести студентов, обучающихся в Сеченовском университете в 2024 году. (n=200).

В зависимости от цели исследования одна и та же статистическая совокупность может рассматриваться и как генеральная, и как выборочная.

Пример
Изучаемая совокупность ― все студенты Сеченовского университета, обучающиеся в 2024 году.

Если цель исследования ― изучить здоровье студентов, которые в 2024 году обучаются в Сеченовском университете, то изучаемая совокупность является генеральной.

Если же цель исследования ― изучить здоровье студентов всех медицинских университетов России, то изучаемая совокупность является выборочной.

Исследователи используют выборочную совокупность, когда изучать генеральную совокупность невозможно или нерационально. Например, когда генеральная совокупность очень велика и изучать ее всю целиком нерационально. Так как такое исследование потребует много времени или средств.

Пример
Необходимо изучить здоровье женщин репродуктивного возраста (от 15 до 49 лет), проживающих в Московской области. Численность женщин этого возраста в данном регионе (объем генеральной совокупности) составляет примерно 2 млн. Чтобы провести медицинский осмотр такого количества женщин, потребуется много времени и врачей. К тому же это довольно затратно для медицинских страховых компаний.

Исследователи формируют выборку из нескольких сотен женщин области (например, n=1000), делают им полноценное обследование, изучают результаты, делают выводы об их здоровье. А затем «переносят» сделанные заключения на всех женщин репродуктивного возраста, проживающих в данном регионе (то есть на всю генеральную совокупность).

Еще выборку используют, когда генеральная совокупность не имеет границ (например, временных).

Пример
Требуется изучить эффективность нового лекарства для снижения артериального давления. Для этого необходимо опробовать его на всех больных, страдающих артериальной гипертензией, проживающих во всём мире. А также всех тех, у кого будет это заболевание в будущем (генеральная совокупность). Выполнить данное исследование невозможно.

Исследователи формируют выборку, которая состоит из нескольких сотен больных артериальной гипертензией (например, n=500), апробируют лекарственное средство на этих пациентах. Затем сделанные выводы относительно эффективности и безопасности лекарства распространяют на всю генеральную совокупность.

Также выборку используют, когда изучение всей генеральной совокупности приведет к ее уничтожению.

Пример
В аптечную сеть с завода производителя поступила партия лекарственного препарата (генеральная совокупность) в форме раствора в герметично закупоренных флаконах. Необходимо оценить качество всей партии (например, N=3000) на предмет концентрации лекарственных веществ в растворе. Для этого необходимо выполнить ряд химических опытов со вскрытием герметичных флаконов и использованием большей части его содержимого.

Естественно, что после такого исследования использовать вскрытый флакон с лекарственным препаратом в лечебных целях нельзя. И если вскрыть и проверить все флаконы партии (то есть изучить всю генеральную совокупность), то будут получены очень точные сведения о качестве изучаемой партии лекарства, но вся партия будет полностью уничтожена.

Поэтому исследователи формируют выборку из нескольких флаконов, вскрывают их, проверяют, и выводы о качестве лекарственного препарата, сделанные по выборке, распространяют на всю партию лекарства (генеральную совокупность).

Возникает логичный вопрос: а можно ли доверять выборочным данным, особенно в медицине? Можно ли утверждать, что, если лекарственное средство показало высокую эффективность для 80 из 100 пациентов выборочной совокупности, то и вся генеральная совокупность больных этим заболеванием будет характеризоваться теми же свойствами? То есть четырёх из каждых пяти пациентов данное лекарство будет эффективно излечивать?

##### Закон больших чисел
Возможность изучать генеральную совокупность по выборке опирается на один из основных законов в теории вероятностей ― закон больших чисел, простейшую схему проявления которого впервые описал швейцарский математик Якоб Бернулли.

Для статистического исследования он звучит так:

1. Результаты выборочного исследования при увеличении числа наблюдений в выборке все больше становятся похожими на истинные характеристики генеральной совокупности.

2. При достижении определенного числа наблюдений в выборке результаты выборочного исследования станут максимально близкими к истинным характеристикам генеральной совокупности.

Другими словами, чем больше объем выборки (чем больше в ней единиц наблюдения), тем точнее исследование и тем больше можно доверять выводам, сделанным при изучении этой выборки. Однако, начиная с некоторого объема наблюдения, точность исследования почти перестает изменяться, и увеличение числа единиц наблюдения перестаёт давать существенное повышение точности. То есть с какого-то момента исследователь начинает делать лишнюю работу, так как он мог бы получить те же самые сведения, изучив меньшее число единиц наблюдения.

Пример
Изучается распространенность табакокурения среди студентов всех медицинских вузов России. Генеральная совокупность достаточно велика, потому решили провести выборочное исследование. Опросили 20 случайно выбранных студентов разных вузов. Все они оказались некурящими.

Можно ли на основании этих данных сделать вывод, что все студенты медицинских вузов страны настолько разумны, что не портят своё здоровье курением? Конечно, нет. Выборка слишком мала, она не способна охарактеризовать всю генеральную совокупность. В неё случайно попали только некурящие студенты.

Если исследователи увеличат объем выборки, то повысится вероятность того, что в ней окажутся курящие студенты. И чем больше будет становиться объем выборки, тем соотношение курящих и некурящих в ней будет ближе к истинному. А при достижении некоторого объема выборки это соотношение будет очень мало отличаться от истинного.

Таким образом, малая выборка не может точно охарактеризовать генеральную совокупность. А всегда ли способна это сделать большая выборка? Нет, не всегда.

Репрезентативность
Обязательным условием выборочной совокупности является репрезентативность (от англ. representative ― представительный), когда все объекты выборки хорошо отражают изучаемые свойства генеральной совокупности.

Выборка должна быть репрезентативной в плане количества и качества.

Количественная репрезентативность связана с объемом выборки и обеспечивает статистическую значимость результатов исследования. Следовательно, чем больше объем выборки, тем более достоверными будут полученные результаты.

Пример количественной репрезентативности можно увидеть в клинических исследованиях, где большое количество пациентов включается в исследование. Это позволяет сделать выводы, применимые к более широкой популяции.

Например, если клиническое исследование эффективности нового препарата включает несколько тысяч пациентов, его результаты будут более надежными и репрезентативными, чем у исследования, в котором участвовали всего несколько десятков пациентов.

Качественная репрезентативность связана с тем, насколько хорошо выборка представляет целевую популяцию. Это значит, что выборка должна быть представительной и разнообразной, чтобы ее характеристики отражали популяцию. Это позволит обобщить результаты исследования на всю целевую группу.

Примером качественной репрезентативности может служить исследование о побочных эффектах лекарства у пожилых людей. Если выборка состоит из представителей разных возрастных групп, включая тех, кто страдает хроническими заболеваниями, проживающих как в городе, так и в сельской местности, то результаты такого исследования будут более качественно репрезентативными и позволят делать более обоснованные выводы о безопасности лекарства для всей группы пожилых пациентов.

Таким образом, в медицине важно не только количество участников исследования, но и их качественное представительство. Чтобы полученные данные были релевантными, достоверными и могли быть применены на практике для улучшения здоровья и качества жизни пациентов.

Существует несколько методов для формирования репрезентативной выборки, представляющих различные подходы к отбору участников исследования:

###### Случайная выборка ― каждый элемент популяции имеет одинаковые шансы попасть в выборку. С помощью случайной выборки можно обеспечить объективность результатов и минимизировать вероятность искажения данных.

###### Стратифицированная выборка ― популяция делится на различные страты (группы) по определенным признакам, а затем из каждой страты отбирается участники выборки. Этот метод позволяет учесть гетерогенность популяции и обеспечить равное представительство различных групп.

###### Кластерная выборка ― популяция делится на кластеры (например, районы, школы, предприятия), из каждого кластера случайным образом выбирается определенное количество участников исследования. Этот метод удобен при работе с большими популяциями и позволяет сократить затраты на сбор информации.

###### Удобственная выборка ― участники выбираются на основе удобства доступа или желания принять участие в исследовании. Этот метод не всегда обеспечивает репрезентативность выборки, так как участники могут существенно отличаться от общей популяции.

###### Квотная выборка ― участники выборки отбираются на основе заранее заданных квот по различным признакам (например, полу, возрасту, образованию). Этот метод позволяет обеспечить баланс по различным характеристикам, однако требует внимательного контроля за процессом отбора участников.

Каждый из этих методов обладает уникальными характеристиками и используется в зависимости от целей исследования, а также от особенностей изучаемой группы и доступных ресурсов для сбора информации. Важно правильно подобрать метод формирования выборки, чтобы обеспечить ее репрезентативность и достоверность результатов исследования.

##### Эффективное проведение статистического исследования

Статистические исследования в медицине играют ключевую роль в обеспечении эффективности и надежности медицинских практик. Они позволяют оценить эффективность лекарственных препаратов, выявить факторы риска развития заболеваний, определить влияние генетических и окружающих факторов на здоровье человека, а также принять обоснованные решения в области здравоохранения.

##### Эффективное проведение статистического исследования в медицине включает в себя несколько ключевых этапов.

1. Определение проблемы исследования

Проблемой в общественном здоровье и здравоохранении могут быть:

недоступность качественной медицинской помощи из-за отсутствия медицинских учреждений, длительных очередей на прием к врачу или недостаточного количества специалистов;
эпидемии и пандемии, требующие масштабных медицинских и предупредительных мер;
проблемы психического здоровья населения;
недостаточное финансирование здравоохранения, приводящее к нехватке медицинского оборудования, медикаментов и низким зарплатам медицинского персонала;
антимикробная резистентность, затрудняющая лечение инфекционных заболеваний и требующая разработки новых подходов к борьбе с этим явлением;
проблемы вакцинации в некоторых регионах, увеличивающие риск возникновения вспышек инфекционных заболеваний, которые могли бы быть предотвращены с помощью прививок.
2. Формулирование задач исследования

На этом этапе определяются цель и задачи исследования, формулируются гипотезы, которые необходимо подтвердить или опровергнуть. Четкость постановки задач обеспечивает правильное планирование и проведение исследования.

3. Определение методологии исследования

Выбираются методы сбора данных, разрабатывается план исследования, определяются критерии включения и исключения пациентов, а также проводится расчет объема выборки.

4. Сбор данных

Проведение статистического исследования включает в себя сбор разнообразной информации о пациентах или группах, которые подвергаются анализу. Этот этап требует тщательного и строго контролируемого сбора данных.

5. Обработка данных

После сбора данных происходит их структурирование, кодирование и ввод в статистические программы для анализа. На этом этапе используются статистические методы для выявления в данных закономерностей и тенденций.

6. Анализ результатов

После обработки данных проводится анализ полученных результатов с использованием статистических методов. На основе анализа делаются выводы и формулируются рекомендации для практического применения.

7. Интерпретация результатов и формулирование выводов

Полученные данные помогают принимать важные решения в области медицины, разрабатывать новые методики лечения, профилактики и диагностики.

##### Пример

Проблема: в некоторых районах города наблюдается высокий уровень заболеваемости астмой среди подростков.

Задача: провести статистическое исследование с целью выявить возможные факторы, влияющие на высокий уровень заболеваемости астмой среди подростков в указанных районах.

Цель: проанализировать связь между экологическими условиями, образом жизни, семейным положением и развитием астмы среди подростков.

Гипотеза: высокие показатели загрязнения окружающей среды, неблагоприятные условия проживания, отсутствие спортивной активности и наличие неблагополучной семейной ситуации способствуют увеличению риска развития астмы среди подростков в определенных районах города.

Определение методологии исследования: выбор методов сбора данных, разработка плана исследования, определение критериев включения и исключения подростков.

Сбор данных: сбор информации о заболеваемости астмой, экологических условиях, образе жизни и семейных факторах у подростков.

Обработка данных: структурирование, кодирование и анализ данных с использованием статистических программ.

Анализ результатов: анализ полученных данных, проверка гипотезы, выявление закономерностей и выводов о влиянии различных факторов на заболеваемость астмой среди подростков в районах города.

Подготовка отчета: формулирование выводов и рекомендаций на основе полученных результатов и подготовка научного отчета для дальнейшего запроса.

Продолжаем углубляться в основные понятия, в этом уроке мы разберем статистические признаки.

##### Единица наблюдения 
― конкретный объект или субъект, на который направлено внимание и изучение в рамках исследования. Это может быть пациент, студент, врач, который становится объектом анализа и изучения для получения информации о его характеристиках, свойствах и поведении.

##### Признаки сходства 
― характеристики или аспекты, общие у различных объектов или единиц наблюдения. Эти признаки позволяют выявить сходство между объектами и установить общие черты или свойства, которые характеризуют их как однородные или схожие, с точки зрения определенных параметров.

##### Признаки различия 
― свойства или характеристики, по которым объекты или единицы наблюдения различаются. Эти признаки помогают выделить индивидуальные особенности и уникальные аспекты каждого объекта, позволяя провести сравнение и выявить различия между ними.

##### Учетные признаки 
― параметры, характеристики или переменные, которые используются для учета и описания объектов или единиц наблюдения. Эти признаки могут быть как качественными, так и количественными. Служат для систематизации и классификации данных, а также для последующего анализа и интерпретации результатов исследования.

Учетные признаки классифицируются по роли в совокупности и характеру.



##### Пример

Единица наблюдения ― москвичи, заболевшие короновирусом.

Признаки сходства ― место проживания (Москва), заболевание (короновирус).

Признаки отличия ― температура, кашель (сухой/влажный), затрудненность дыхания (наличие/отсутствие), шумы в легких (наличие/отсутствие) и так далее.



##### Классификация признаков по характеру: качественные, количественные

##### Количественные признаки 
представляют собой числовые значения, которые могут быть измерены и подвергнуты математическим операциям. О количественных мы можем сказать: «на три единицы меньше», «в шесть раз больше». Они позволяют более точно описывать и анализировать данные, поскольку представляют собой конкретные численные значения. 

Количественные признаки могут быть непрерывными или дискретными.

###### Непрерывные признаки 
могут принимать любые значения в определенном диапазоне. Например, температура, рост пациента, биохимические показатели крови. Над непрерывными признаками можно проводить арифметические операции.

###### Дискретные признаки
, напротив, принимают конкретные значения, такие как количество детей в семье, число рецидивов, число заболеваний у одного больного, число вызовов скорой помощи, число пациентов, поступающих в приемное отделение городской больницы.

Использование количественных признаков позволяет проводить более глубокий анализ данных, выявлять закономерности, строить математические модели и делать прогнозы.

##### Качественные признаки (категориальные)
 ― свойства объектов, которые характеризуются не числовыми значениями, а категориями, метками или описательными обозначениями. Они описывают качественные или *категориальные* характеристики объекта и определены через их наличие или отсутствие, принадлежность к определенной группе или категории.

Качественные признаки позволяют классифицировать объекты по определенным свойствам и качествам, что позволяет исследователям делать выводы о сходствах и различиях между объектами в категориальном аспекте.

Например, способ лечения: хирургический, радиотерапия, обсервационный. Такие признаки ещё называют категориальными, поскольку их можно отнести к той или иной категории.

##### Среди категориальных признаков можно выделить:

###### Номинальные признаки 
― это переменные, значения которых не имеют внутреннего порядка, а просто обозначают различные категории или качественные характеристики. В медицине номинальные признаки играют важную роль при классификации пациентов, диагнозе заболеваний и определении других особенностей.

Примеры номинальных признаков в медицине можно найти в различных областях, таких как эпидемиология, диагностика, лечение и мониторинг пациентов:

###### Тип заболевания 
― инфекционное, онкологическое, аутоиммунное или другое. Номинальный признак «тип заболевания» помогает врачам классифицировать и выбирать соответствующие методы лечения.

###### Группа крови человека
― номинальный признак, который может быть полезен для применения при трансфузии крови, определении совместимости доноров и реципиентов, а также в генетических исследованиях.
###### Тип вакцинации 
― при проведении вакцинации определяется тип вакцины, которая была использована (вирусная, бактериальная, деленная, живая или мертвая). Этот номинальный признак важен для отслеживания эффективности вакцинации и реакции пациента на прививку.
###### Пол пациента 
― часто используется как номинальный признак для сегментации пациентов по половому признаку. Это может быть важно при анализе различий в заболеваниях, реакции на лечение и других параметрах.
###### Методы диагностики 
― рентгенологические, лабораторные, исследования с использованием ультразвука и томографии. Это помогает идентифицировать, какие методы были использованы для диагностики конкретного заболевания.
##### Дихотомические (бинарные) 
признаки представляют собой категориальные переменные, которые могут принимать только два возможных значения. Такие признаки делят объекты на две группы в зависимости от наличия или отсутствия определенного свойства.

##### Пример
Дихотомический признак «курение» может принимать два значения: «да» или «нет». Этот признак является ключевым при анализе здоровья пациентов, так как курение сильно влияет на развитие многих заболеваний. Наличие или отсутствие этой привычки может быть решающим фактором при определении риска развития сердечно-сосудистых заболеваний, рака легких, хронического обструктивного заболевания легких и других патологий.

Данный дихотомический признак «курение» позволяет проводить анализы, исследовать влияние курения на состояние здоровья пациентов, разрабатывать профилактические и лечебные мероприятия с учетом данного фактора.

Номинальные признаки в медицине играют важную роль в классификации и анализе информации о здоровье пациентов. Понимание и правильное использование таких признаков помогает медицинским специалистам принимать более компетентные решения при диагностике и лечении различных заболеваний.

##### Порядковые (ранговые, ординальные) признаки 
представляют собой переменные, значения которых имеют внутренний порядок, но не обязательно равные интервалы между ними. Такие признаки позволяют упорядочить объекты по степени или уровню наблюдаемого свойства, но не позволяют проводить математические операции с точными числовыми значениями. Порядковые признаки широко используются в медицине для оценки тяжести заболеваний, уровня боли, степени нарушения функций органов и других параметров:

###### Шкала боли 
(нумерическая шкала оценки боли от 1 до 10). Врачи могут попросить пациента оценить свою боль по шкале от 1 (без боли) до 10 (наивысшая возможная боль). Такая шкала позволяет оценить интенсивность болевого синдрома и принять решение о необходимом лечении.
###### Стадии заболевания (например, стадии рака). 
Часто заболевания классифицируются на различные стадии в зависимости от распространенности процесса. Например, рак может иметь стадии от I до IV, где более высокая стадия указывает на более продвинутый и тяжелый случай заболевания. Такая классификация помогает врачам определить тактику лечения и прогноз для пациента.
###### Уровни образования (например, начальное, среднее, высшее). 
Данный порядковый признак может использоваться при исследованиях в области общественного здравоохранения для анализа влияния образования на здоровье пациентов. Уровень образования может быть связан с доступом к медицинской информации, привычками ведения здорового образа жизни и другими факторами, влияющими на состояние здоровья.



Использование порядковых признаков в медицинских исследованиях позволяет более точно оценить различные характеристики и свойства пациентов, что способствует выработке эффективных стратегий диагностики, лечения и профилактики различных заболеваний.

##### Классификация признаков по роли в совокупности: факторные, результативные

По роли в совокупности выделяют факторные признаки ― переменные, которые могут влиять на результативный признак. То есть на конечный результат или исход, который мы хотим изучить или прогнозировать. Таким образом, факторные признаки имеют причинно-следственную связь с результативными показателями.


##### Пример


1. Факторный признак: уровень холестерина в крови.

Результативный признак: риск развития сердечно-сосудистых заболеваний.

В данном случае уровень холестерина в крови является факторным признаком, который может влиять на риск развития сердечно-сосудистых заболеваний. Повышенные значения холестерина могут быть связаны с развитием атеросклероза и увеличением вероятности сердечного приступа или инсульта.

2. Факторный признак: степень накопления токсинов в организме.

Результативный признак: функция почек.

Высокий уровень токсинов в организме может негативно сказываться на функции почек, приводя к их дисфункции или хронической почечной недостаточности. В данном случае степень накопления токсинов является факторным признаком, который может влиять на состояние функции почек как результативный признак.

3. Факторный признак: уровень физической активности.

Результативный признак: изменение веса пациента.

Уровень физической активности может влиять на изменение веса пациента. Недостаточная физическая активность часто связана с лишним весом и ожирением, в то время как активный образ жизни способствует снижению веса и улучшению общего состояния здоровья.

Исследование факторных и результативных признаков в медицине позволяет выявлять взаимосвязи между различными параметрами здоровья пациентов и определять наиболее значимые факторы, которые влияют на исход заболеваний или эффективность лечения. Это помогает разрабатывать индивидуализированные подходы к диагностике, лечению и профилактике заболеваний с учетом особенностей каждого конкретного случая.

Пример
В исследовании единицей наблюдения является студент, который учится в Сеченовском университете на протяжении всего периода обучения.

Признаки, относящиеся к качественным характеристикам, включают в себя такие факторы, как пол, наличие зависимости от курения, общее состояние здоровья.

Количественные признаки включают в себя параметры, такие как возраст, рост, количество сигарет, выкуриваемых в день, продолжительность заболевания, стаж курения.

Факторными признаками являются наличие вредных привычек и стаж курения.

А результативными ― наличие заболевания, состояние здоровья.

#### _4. Абсолютные и относительные величины. Интенсивный показатель_

---
<span style="font-size:0.8em;"> [TOC](#toc) </span>

Описательная статистика в медицине играет ключевую роль в анализе и интерпретации данных. Этот метод позволяет систематизировать, описывать и представлять информацию о популяции пациентов, их заболеваниях, лечении и других медицинских параметрах.

##### Описательная статистика имеет несколько ключевых задач:

1. Описание и классификация данных

Включает в себя вычисление мер центральной тенденции (среднее, медиана, мода) и мер изменчивости (стандартное отклонение, диапазон, квартили).

2. Визуализация данных

Создание графиков, диаграмм и других визуальных представлений данных. Это помогает выявить закономерности, аномалии и взаимосвязи между переменными.

3. Сводка данных и выявление основных характеристик

Сведение данных в краткую форму, выделение основных характеристик выборки или популяции помогает исследователям быстрее понимать сущность данных и делать выводы.

4. Сравнение групп и распределений

Описательная статистика позволяет сравнивать различные группы пациентов или факторы в выборке для выявления статистически значимых различий. Это особенно важно при проведении медицинских исследований и анализе эффективности лечения.

5. Выявление ассоциаций и корреляций

Выявление связи и корреляции между переменными позволяет понять влияние одних факторов на другие. Это полезно при исследовании медицинских данных и их взаимосвязей.

В процессе анализа медицинских данных используются абсолютные и относительные величины.

##### Абсолютные величины
используются для описания фактических числовых значений данных, например, количество пациентов с определенным диагнозом или число лекарственных препаратов, выписанных в определенный период.

##### Относительные величины 
позволяют сравнивать данные в различных контекстах и на разных уровнях. Например, относительное количество пациентов с определенным заболеванием может быть выражено в процентном соотношении к общему количеству пациентов в клинике. Это помогает оценить распространенность заболевания и эффективность лечения.


##### Различают четыре основных вида относительных величин:

###### Экстенсивные величины
, такие как объем крови, масса тела, количество кровяных клеток, зависят от объема организма.
Например, общая масса крови в организме человека является экстенсивной величиной, так как ее значение изменяется в зависимости от общего объема кровообращения в организме.

###### Соотношения 
позволяют выявлять взаимосвязи между различными параметрами организма.
Например, отношение количества лейкоцитов к общему числу клеток крови может указать на наличие воспалительного или инфекционного процесса в организме. Такие соотношения помогают диагностировать различные заболевания и контролировать их лечение.

###### Наглядные величины 
в медицине, такие как индекс массы тела, коэффициенты роста и веса, служат для визуализации и оценки физического состояния человека.
Например, использование индекса массы тела позволяет оценить соответствие между массой тела и ростом человека, что важно для определения возможного риска ожирения или дефицита веса.

###### Интенсивный показатель ― 
ключевой инструмент для диагностики различных состояний здоровья человека. Этот тип показателя обозначает характеристику, которая не зависит от общего объема организма, а отражает уровень определенного физиологического процесса. В медицине интенсивные показатели могут быть использованы для оценки работы органов и систем, контроля за состоянием пациента, а также для мониторинга эффективности лечения.
Например, пульс, давление, температура тела ― это типичные интенсивные показатели, которые активно используются в клинической практике. Измерение пульса позволяет оценить работу сердечно-сосудистой системы, выявить аритмии, недостаточность кровообращения и другие заболевания. Анализ показателей давления помогает контролировать работу сердца, обнаруживать гипертонию, гипотонию. Температура тела также является важным интенсивным показателем, позволяющим определить наличие воспалительных процессов, инфекций или других патологий.

Для расчета интенсивного показателя используются специальные формулы, алгоритмы и методики, зависящие от конкретного параметра, который необходимо оценить. Например, для расчета среднего артериального давления необходимо провести измерения верхнего и нижнего артериального давления, затем применить математическую формулу для определения среднего значения.

Другой пример расчета интенсивного показателя ― это индекс массы тела (ИМТ), который позволяет оценить соотношение массы и роста человека. Для расчета ИМТ необходимо разделить массу (в килограммах) на квадрат роста (в метрах). Полученное значение ИМТ помогает определить степень ожирения или дефицита массы тела у пациента.

##### Пример
Расчет интенсивного показателя «Сердечные обороты в минуту»:

1. Измерьте пульс у пациента. Обычно измеряется на сонной артерии на шее или на лучевой артерии на запястье.

2. Запишите количество пульсов за 1 минуту. Считайте количество ударов сердца в течение 60 секунд.

3. Рассчитайте сердечные обороты в минуту (частота сердечных сокращений). Количество пульсов, подсчитанное за 1 минуту, и будет значением сердечных оборотов в минуту.

Если за 1 минуту у пациента было 72 удара сердца, то сердечные обороты в минуту (сердечный ритм) равны 72 ударам в минуту. Этот показатель может помочь врачам определить наличие аритмий, скрытых заболеваний сердца и оценить общее состояние сердечно-сосудистой системы.

Расчет интенсивных показателей в медицине требует точности, аккуратности и использования специализированного оборудования и методов. Корректный расчет интенсивных показателей позволяет точно определить состояние здоровья и эффективность проводимого лечения, что важно для обеспечения качественной медицинской помощи.

#### _5. Способы графического изображения интенсивных показателей_

---
<span style="font-size:0.8em;"> [TOC](#toc) </span>

Рассмотрим несколько способов графического представления интенсивных показателей:

##### Линейные графики 
могут быть использованы для отображения динамики изменения показателей во времени. На оси X откладывается время, а на оси Y ― значения интенсивных показателей. Такие графики позволяют наглядно показать тренды и колебания.
##### Столбчатые диаграммы 
могут использоваться для сравнения разных интенсивных показателей между собой или для представления значений показателей в различных условиях или ситуациях. Каждый столбец соответствует определенному показателю.
##### Круговые диаграммы
могут быть полезны для отображения долей различных составляющих интенсивных показателей. Например, показать процентное соотношение разных компонентов в целом значении показателя.
##### Интерактивные графики
, созданные с использованием современных технологий, позволяют дополнительно анализировать данные по интенсивным показателям. Это может включать в себя навигацию по временной шкале, подробный анализ конкретных точек на графике и другие функции.
##### Тепловые карты 
могут использоваться для визуализации пространственного распределения интенсивных показателей. Они позволяют быстро выявить зоны с высокими или низкими значениями показателей на основе цветовой градации.
Графическое представление интенсивных показателей помогает лучше понять данные, сделать выводы и принять информированные решения в области медицины, физиологии или спорта.

Линейная диаграмма
Представим графическое отображение изменения уровня кислорода в крови (SpO2) у пациента в течение суток в виде линейного графика.

На горизонтальной оси (ось X) откладывается время с интервалом в часы или минуты, а на вертикальной оси (ось Y) ― уровень насыщения кислорода в крови в процентах от 0 до 100%.

График показывает динамику изменения уровня SpO2 в различное время суток. Например, в начале дня уровень кислорода в крови может быть стабильным на определенном уровне, затем наблюдается некоторое понижение в период активности пациента, после чего возвращается к норме во время отдыха и сна.

Анализируя такой график, врач может выявить периоды, когда уровень кислорода в крови пациента ниже оптимального значения. И принять соответствующие меры для повышения его показателей.

Линейный график изменения уровня кислорода в крови является одним из ключевых инструментов для отслеживания состояния пациента и помогает врачам принимать решения на основе объективных данных.

Радиальная диаграмма
Визуальное представление данных, в котором информация представлена в виде кругов, расходящихся от центральной точки. Обычно используется для отображения частоты или интенсивности встречаемости определенных значений или категорий в зависимости от расстояния от центра. Такая диаграмма позволяет наглядно представить распределение данных и выявить закономерности или тенденции. Радиальные диаграммы могут быть полезны для анализа больших объемов информации и выявления основных трендов или аномалий.

В медицине радиальные диаграммы используют для отслеживания динамики изменения клинических показателей у пациентов и сравнения эффективности лечения. Или для исследования суточных колебаний числа вызовов скорой помощи, колебаний числа выписываемых и госпитализируемых в стационары больных по дням недели.

Также радиальная диаграмма может отобразить уровень различных биохимических показателей в крови (например, глюкозы, холестерина, белка) до и после применения определенного лекарственного препарата или процедуры. По такой диаграмме можно оценить, как показатели меняются под воздействием терапии.

Другой пример применения радиальных диаграмм в медицине ― оценка состояния здоровья сердечно-сосудистой системы. На диаграмме можно отобразить такие показатели, как артериальное давление, частота пульса, уровень холестерина, индексы кровообращения и другие параметры, чтобы визуально оценить общее состояние сердечно-сосудистой системы пациента.

Преимущества радиальных диаграмм:

возможность визуализации данных с несколькими измерениями;
наглядное представление относительных значений;
удобство сравнения различных параметров.
Кроме того, радиальные диаграммы достаточно легко интерпретировать, что делает их эффективным инструментом для анализа сложных данных.

Визуальное представление данных в виде вертикальных столбцов. Их длина пропорциональна количественному показателю или значениям переменной, которую они представляют. Столбчатые диаграммы облегчают восприятие информации и выявление тенденций, а также позволяют наглядно выделить ключевые аспекты анализируемых данных.

#### _6. Экстенсивный показатель_

---
<span style="font-size:0.8em;"> [TOC](#toc) </span>

Экстенсивный показатель — относительная величина, показывающая, как велика отдельная часть по отношению ко всей изучаемой совокупности. Одним из основных экстенсивных показателей, используемых в медицинских исследованиях, является общее количество заболевших в определенной популяции или общее количество случаев заболевания в определенном временном периоде.



Экстенсивные показатели могут быть также связаны с оценкой распространенности определенных заболеваний или сберегающих здоровье практик. Например, показатели, отражающие количество диагностированных случаев редких заболеваний в определенной популяции, помогут понять необходимость создания специализированных программ по их обнаружению и лечению.

Экстенсивные показатели также могут быть использованы для анализа эффективности профилактических программ или лечебных методов. На основе данных о количестве проведенных процедур, выписанных лекарств или успешно пройденных курсов реабилитации можно делать выводы о том, насколько эффективны эти меры для улучшения состояния пациентов и снижения заболеваемости.

Таким образом, экстенсивные показатели в медицине играют значительную роль в оценке здоровья населения, планировании медицинских программ и формировании стратегий по улучшению общественного здоровья.

##### Классификация экстенсивных показателей:

###### Показатели удельного веса части в целом. 
Например, доля заболевших короновирусом среди всех заболеваний.


###### Показатели структуры или распределения. 
Распределение всех заболеваний, зафиксированных за год, по отдельным заболеваниям.

###### Расчет экстенсивных показателей:

1. Процент заболеваемости в определенной популяции

Для определения процента заболеваемости в популяции можно рассчитать отношение числа заболевших к общему числу людей в этой популяции и умножить на 100%.

Например, если в популяции из 1000 человек было 50 заболевших, то процент заболеваемости составит (50 / 1000) * 100% = 5%.

2. Общее количество случаев заболеваний

Для определения экстенсивного показателя общего количества случаев заболеваний в популяции за определенный период времени нужно просто сложить все единичные значения количества заболевших за каждый месяц или год.

Например, если у нас есть данные о числе заболевших в популяции каждый месяц за год, то, суммируя все значения, мы получим общее количество случаев заболеваний за весь год.

3. Суммарное количество проведенных операций

Для оценки экстенсивного показателя суммарного количества проведенных операций в больнице за год можно сложить все значения операций за каждый месяц.

Например, если в январе было проведено 100 операций, в феврале 120 операций и так далее, то общее количество операций за год будет равно сумме этих значений.

#### _7. Способы графического изображения экстенсивных показателей_

---
<span style="font-size:0.8em;"> [TOC](#toc) </span>

Графически экстенсивный показатель можно изобразить только в виде секторной (круговой) или внутристолбиковой диаграммы. Рассмотрим правила построения таких диаграмм на примере с инфекционными заболеваниями из предыдущего урока.

Алгоритм построения секторной диаграммы:

1. Определение данных

Начните с определения данных, которые вы хотите представить на диаграмме. Это может быть распределение чего-либо по категориям или доли чего-то в общем объеме.

2. Расчет процентных долей

Для каждой категории вычислите ее процентное соотношение относительно общего объема данных. Проценты должны быть представлены в виде долей, равных дроби от 1. Например, 0.25, 0.5, 0.75.

3. Выбор порядка категорий

Установите порядок категорий, который будет отображен на диаграмме. Обычно порядок идет по часовой стрелке, от самого большого сектора к самому маленькому.

4. Рассчитайте углы секторов

Для каждой категории вычислите угол сектора, который соответствует ее процентной доле от общего угла круга (360 градусов). Угол можно рассчитать, умножив процент на 360.

5. Нарисуйте диаграмму

Используя углы секторов и порядок категорий, нарисуйте круговую диаграмму с соответствующими секторами для каждой категории. Убедитесь, что каждый сектор четко отражает свою процентную долю.

6. Добавьте легенду

Для большей понятности добавьте легенду, которая поясняет категории или данные, представленные на диаграмме.

Характеризует соотношение между двумя совокупностями, не связанными между собой (соотношение абортов и родов, обеспеченность населения койками, врачами, соотношение врачей и другого медицинского персонала).

Показатели описательной статистики можно разбить на три основные группы:

##### Меры центральной тенденции 
― статистические показатели, которые используются для описания среднего значения или «средней точки» распределения данных. Они представляют собой характеристики, которые позволяют суммировать или усреднять данные в выборке для представления их общего положения или среднего значения. Меры центральной тенденции помогают сжать разнообразные данные в компактный и информативный числовой показатель, который отражает основные характеристики распределения и позволяет делать выводы о типичном или среднем значении в исследуемой выборке.

###### Меры центральной тенденции: среднее, мода, медиана.

##### Меры разброса (изменчивости)

Если в наборе данных нет однозначно выраженного модального значения (если все значения встречаются одинаковое количество раз), то такой набор данных называется

###### 2. Исследования на популяционном уровне

В здравоохранении мода и медиана могут применяться для анализа результатов исследований, таких как исследования заболеваемости, смертности или эффективности лечения. Например, определение моды возраста больных с определенным заболеванием может помочь в планировании мероприятий профилактики и лечения.

###### 3. Оценка качества медицинских услуг

Медиана может использоваться для оценки времени, требуемого на ожидание приема в клинике или скорую помощь. Мода может показать наиболее часто встречающиеся жалобы пациентов, что поможет улучшить качество медицинских услуг.

###### 4. Оценка лабораторных показателей

Мода и медиана могут быть применены для анализа лабораторных данных, например, уровня гемоглобина, холестерина или других показателей здоровья. Эти статистические показатели могут помочь в выявлении аномалий и требовать дополнительного вмешательства или наблюдения.



##### Гистограмма 
― график, показывающий, как часто значение переменной встречается на определенном промежутке. По оси абсцисс откладывается значение переменной, а по оси ординат указывается, как часто значение этой переменной встречается на определенном интервале. Интервалы можно выбрать разной длины.

Если интервалы будут слишком большими, то гистограмма будет грубой и малоинформативной, если интервалы очень малы, то гистограмма будет очень разреженной.

##### Замечания 

Среднее арифметическое зависит от всех значений в наборе данных, включая крайние значения, которые могут быть нехарактерными для общего объема данных. Поэтому при анализе данных сведения о среднем арифметическом часто дополняются указанием медианы.

Среднее арифметическое, мода и медиана по-разному характеризуют данные, полученные в результате наблюдений. При анализе данных в зависимости от ситуации можно использовать любой из этих показателей, комбинацию двух или даже все три.

Если распределение унимодально, симметрично и не имеет заметных выбросов, то можно использовать любую меру центральной тенденции, так как все меры имеют приблизительно одинаковые значения.

Если распределение имеет несколько мод, ассиметрично или имеет заметные выбросы, то использование в анализе данных среднего значения может привести к некорректным результатам. В таком случае для того, чтобы охарактеризовать данные с точки зрения степени выраженности некоторого количественного признака целесообразнее использовать моду и/или медиану.

Среднеквадратическое отклонение и дисперсия неустойчивы к выбросам, как и среднее арифметическое.

##### Стандартное (среднеквадратическое) отклонение 
измеряет разброс значений в выборке относительно их среднего значения (в тех же единицах измерения, что и сама переменная) и позволяет оценить степень изменчивости данных. Чем больше значение стандартного отклонения, тем больший разброс данных можно наблюдать.


В медицинской статистике стандартное отклонение часто используется для оценки изменчивости показателей здоровья, результатов лабораторных исследований, эффективности лечения, а также для сравнения групп пациентов и оценки рисков.

Например, при изучении результатов клинических испытаний нового лекарства стандартное отклонение позволяет оценить, насколько значимы отклонения пациентов от среднего значения эффективности лечения. Это помогает сделать выводы о безопасности и эффективности применения препарата.

Стандартное отклонение также используется в медицинских исследованиях для проверки гипотез, определения доверительных интервалов, анализа динамики заболеваний и тенденций здоровья населения. Например, при изучении распространенности сердечно-сосудистых заболеваний в определенной популяции стандартное отклонение может помочь определить степень разброса данных о состоянии здоровья и выявить факторы риска.

##### Дисперсия

Это мера разброса или вариации значений в наборе данных относительно их среднего значения. Вычисляется нахождением среднего квадратического отклонения от среднего значения.


Иными словами, дисперсия показывает, насколько данные распределены вокруг среднего значения. Чем больше дисперсия, тем больше разброс значений относительно среднего, а чем меньше дисперсия, тем ближе данные к среднему значению.

Дисперсия может использоваться для анализа вариабельности различных параметров здоровья и результатов исследований. Например, при изучении эффективности лекарственного препарата или эффективности терапии дисперсия может помочь определить, насколько данные однородны или различны у пациентов.

Также дисперсия может применяться для оценки точности измерений и результатов исследований. Понимание степени разброса данных нужно, чтобы делать правильные выводы и принимать обоснованные решения.

##### Гипотеза 
— это предположение или теоретическое представление о взаимосвязи между явлениями или воздействиями, которое требует проверки на основе фактических данных и экспериментов.

Формулирование гипотез помогает исследователям ориентироваться в изучаемой проблеме, структурировать процесс исследования, а также делать выводы на основе полученных результатов. Кроме того, формулирование гипотез в медицинских исследованиях способствует научной обоснованности, помогая избегать предвзятости при интерпретации результатов. Это повышает объективность выводов и помогает в разработке новых методов лечения и профилактики заболеваний.

Пытаясь решить проблему родильной горячки, молодой венгерский врач Игнац Земмельвейс проверил самые популярные в 19 веке гипотезы. Приглашал священника, который с колоколом обходил палаты, менял порядок раздачи лекарств и диету.

После множества попыток Земмельвейс пришёл к выводу, что горячка — инфекция, и бороться с ней нужно с помощью асептики.

Работа с гипотезами побуждает ученых к тщательному анализу данных, обсуждению идей с коллегами, проведению экспериментов и поиску новых подтверждений или опровержений сформулированных предположений.

###### Основные цели формулирования гипотез в медицине:

исследование причин возникновения заболеваний,

разработка новых методов лечения,

предупреждение распространения заболеваний.

Гипотезы должны быть четко сформулированными, проверяемыми, опровергаемыми и иметь потенциал для дальнейшего развития и корректировки на основе новых данных. Они могут быть как конкретными, основанными на научных данных, так и гипотетическими, открывающими новые пути исследования.

Формулируются гипотезы в начале исследования. Это необходимо, чтобы понимать, какие доказательства нужно собрать для их подтверждения или опровержения. Формулируется две гипотезы: нулевая (H0) и альтернативная (H1).

###### Нулевая гипотеза (H0) 
— утверждение, которое мы стремимся опровергнуть. Обычно H0 формулируется как утверждение о том, что нет разницы или эффекта.

###### Альтернативная гипотеза (H1) 
— утверждение, которое мы стремимся поддержать. Обычно H1 формулируется как утверждение о том, что существует разница или эффект.

То есть нулевая гипотеза — гипотеза об отсутствии различий, а альтернативная гипотеза — о значимости различий.

###### Принципы формулирования гипотез

Гипотезы должны быть конкретными и проверяемыми.

Гипотезы должны быть сформулированы в терминах наблюдаемых переменных.

Нулевая гипотеза должна быть исчерпывающей. То есть должна охватывать все возможные результаты, кроме тех, которые поддерживают альтернативную гипотезу.

Альтернативная гипотеза должна быть исключающей. То есть должна исключать все возможные результаты, которые поддерживают нулевую гипотезу.

###### Пример

Нулевая гипотеза (H0): новое лекарство не более эффективно в снижении артериального давления у пациентов с гипертонией, чем плацебо.

Альтернативная гипотеза (H1): новое лекарство более эффективно в снижении артериального давления у пациентов с гипертонией, чем плацебо.

Если результаты исследования покажут, что новое лекарство статистически значимо более эффективно, чем плацебо, то мы отвергнем нулевую гипотезу и поддержим альтернативную. Это будет означать, что новое лекарство, вероятно, более эффективно в снижении артериального давления, чем плацебо.

#### _3. Статистический критерий. Уровень статистической значимости_

---
<span style="font-size:0.8em;"> [TOC](#mod1) </span>

##### Статистический критерий 
— числовое значение или правило, используемое для принятия статистических решений на основе данных из выборки. Позволяет оценить вероятность того, что наблюдаемые различия или эффект случайны, а также помогает определить, являются ли полученные результаты статистически значимыми.

Этот инструмент помогает исследователям делать обоснованные выводы на основе статистического анализа данных и проверять гипотезы, касающиеся взаимосвязей или различий в исследуемых явлениях.

##### Два основных типа методов, используемых для проведения статистического анализа данных:

параметрические,

непараметрические.

Выбор между параметрическими и непараметрическими критериями зависит от характера данных и выполнения предположений.


###### Параметрические критерии
, такие как t-критерий Стьюдента или дисперсионный анализ, предполагают, что данные распределены нормально и имеют определенные параметры. Они обычно более мощные, чем непараметрические критерии, и могут обеспечить более точные результаты при выполнении всех предпосылок.

###### Непараметрические критерии
, такие как критерий Уилкоксона-Манна-Уитни или критерий Краскела-Уоллиса, являются более гибкими. Могут быть использованы в случаях, когда данные не удовлетворяют предположениям о нормальном распределении или равенстве дисперсий. Они менее чувствительны к отклонениям от стандартных условий и могут быть применены к нестандартным или категориальным данным.



##### Уровень статистической значимости 
— обозначается как p-значение, это вероятностный порог, который используется в статистическом анализе для оценки степени, до какого предела результаты исследования могут быть случайными. Он отражает вероятность того, что различия или связи между данными считаются статистически значимыми.


Чем более низкий уровень статистической значимости, тем менее вероятно, что наблюдаемые различия или связи могут быть объяснены случайными факторами. И тем больше оснований считать результаты исследования достоверными.

P-значение измеряет вероятность столь же или более экстремальных результатов, чем те, что были получены в выборке. При условии справедливости нулевой гипотезы (гипотезы о равенстве средних, отсутствии эффекта и так далее). Чем меньше p-значение, тем меньше вероятность того, что наблюдаемые данные могут быть случайными. И тем более статистически значимым считается результат.

В медицине уровень статистической значимости играет важную роль при оценке эффективности лекарственных препаратов, результатов клинических исследований, оценке обработки данных пациентов и многих других областях. Определение p-значения позволяет ученым и врачам оценивать достоверность полученных результатов и делать выводы о статистической значимости эффекта или различий между группами.

Важно помнить, что p-значение само по себе не говорит о практической значимости результатов и не определяет их клиническую важность.

Даже очень маленькое p-значение может быть неинформативным, если размер эффекта маленький или клиническое значение результатов незначительно. Таким образом, анализ p-значения следует проводить вместе с другими статистическими показателями и контекстуализировать с учетом клинической релевантности.

Большинство прикладных исследователей в медицине используют низший уровень статистической значимости, чтобы отклонить нулевую гипотезу H0:

p ≤ 0,05

Это означает, что различия достоверны или существенны на 5% уровне значимости. То есть вероятность того, что различия незначимы (несущественны, недостоверны) — не больше 5%.

Реже в медицинской статистике применяется достаточный уровень значимости — 1% (р≤0,01) и высший — 0,1% (р≤0,001).


Важно понимать, что статистическая значимость, хотя и является важным инструментом в анализе данных, не всегда означает истинное значение или причинно-следственную связь.

Статистическая значимость может быть искусственной из-за скрытых или вмешивающихся факторов, которые искажают результаты и ведут к неверным выводам. Более того, статистическая значимость как таковая не предоставляет информации о клинической значимости или практической важности полученных результатов. Важно учитывать контекст и осмысленность результатов, чтобы корректно интерпретировать статистические выводы и принимать обоснованные решения на их основе.

#### _2. Классификации видов гипотез_

---
<span style="font-size:0.8em;"> [TOC](#mod1) </span>

##### По содержанию

###### гипотезы о законах распределения

Н0: распределение длительности восстановления после операции у пациентов с ранением наружных геморроидальных узлов подчиняется экспоненциальному закону распределения.

Предположим, что у пациентов после операции измерили время, необходимое для полного восстановления здоровья (в днях). Гипотеза утверждает, что это время следует случайному, непрерывному экспоненциальному распределению. И что длительность восстановления у разных пациентов независима друг от друга.

Для проверки этой гипотезы можно собрать данные о времени восстановления у группы пациентов, а затем применить статистические методы, такие как тесты согласия. Чтобы увидеть, насколько хорошо данные согласуются с предполагаемым экспоненциальным распределением. В случае подтверждения гипотезы это может помочь оптимизировать схемы лечения и точнее предсказывать период восстановления после операции.

Таким образом, гипотеза о законе распределения времени восстановления после операции у пациентов с ранением наружных геморроидальных узлов представляет собой интересное направление для исследований в медицине и может принести ощутимую пользу практикующим врачам и пациентам.

###### гипотезы о параметрах распределения

Н0: параметры распределения уровня определенного биомаркера у пациентов с диагностированным раком не изменятся после применения нового противоопухолевого препарата.

Уровень биомаркера, связанный с прогрессированием ракового заболевания, будет снижаться у большинства пациентов после лечения новым препаратом.

Эта гипотеза подразумевает, что новый противоопухолевый препарат не может оказать воздействие на уровень определенного биомаркера, который играет ключевую роль в прогрессировании рака. Исследование параметров распределения этого биомаркера до и после лечения может помочь установить эффективность нового препарата и оценить его потенциальное влияние на течение заболевания у пациентов.

##### Гипотезы о средних

###### о значимости различия между средними

Н0: средняя концентрация определенного белка в крови у пациентов с различными типами сахарного диабета не отличается.

Мы предполагаем, что у пациентов с диабетом типа 1 (инсулинозависимый сахарный диабет) средняя концентрация этого белка в крови будет значительно отличаться от пациентов с диабетом типа 2 (инсулиннезависимый сахарный диабет). И что параметры распределения концентрации этого белка будут различны в зависимости от типа диабета.

Для проверки этой гипотезы можно собрать данные о концентрации белка в крови у обеих групп пациентов и сравнить их средние значения. Далее можно использовать статистические методы, такие как t-тест (подробно рассмотрим его в следующих модулях), для определения статистической значимости различий между этими параметрами. Если различия будут статистически значимыми, это может указывать на важность учета типа диабета при диагностике и лечении пациентов.

Таким образом, гипотеза о параметрах распределения концентрации белка в крови у пациентов с различными типами сахарного диабета может помочь лучше понять особенности заболевания и оптимизировать подходы к лечению каждой его формы.

###### о равенстве среднего определенному значению

H0: среднее количество мочевины в моче у пациентов с хронической почечной недостаточностью в возрасте старше 60 лет превышает среднее количество у здоровых людей того же возраста.

Для проверки этой гипотезы исследователям необходимо собрать выборки мочи у обеих групп. Провести анализ содержания мочевины и сравнить их средние значения. Используя статистические методы, такие как дисперсионный анализ или t-критерий Стьюдента (подробно рассмотрим их в следующих модулях), исследователи могут определить наличие статистически значимых различий между средними значениями мочевины в моче у этих двух групп.

Если среднее количество мочевины у пациентов с хронической почечной недостаточностью действительно превышает среднее у здоровых людей, это указывает на необходимость контроля уровня мочевины у пациентов с почечными заболеваниями для эффективного управления их состоянием.

##### Гипотезы о дисперсиях

###### о равенстве дисперсий определенному значению

H0: дисперсия давления у пациентов с гипертонией равна 80 мм ртутного столба.

Для проверки этой гипотезы исследователи могли бы собрать данные о давлении у группы пациентов с гипертонией. И провести анализ, чтобы узнать, насколько близка дисперсия этих значений к предполагаемому значению 80 мм ртутного столба.

Если статистические методы покажут, что дисперсия давления у пациентов с гипертонией действительно равна или близка к 80 мм ртутного столба, это может иметь практическое значение для мониторинга здоровья и эффективности лечения у пациентов с подобными давлением.

###### о значимости различия дисперсий двух совокупностей

H0: дисперсия реакции пациентов на новое средство для лечения гипертонии будет различаться в зависимости от их возрастной группы.

Речь идёт о том, что у пациентов разных возрастных групп разброс в реакции на новый препарат может быть различным. Для проверки этой гипотезы исследователи могут собрать данные о результате лечения гипертонии с использованием нового средства у пациентов различных возрастов. Затем оценить дисперсию реакции на препарат в каждой возрастной группе и сравнить их.

Используя анализ дисперсий (ANOVA) или другие статистические методы, исследователи могут определить, есть ли статистически значимые различия в дисперсии реакции на препарат между разными возрастными группами. Подтверждение гипотезы говорит, что возраст является фактором, влияющим на вариабельность эффективности препарата у пациентов с гипертонией. Это знание может помочь персонализировать лечение по возрасту пациентов.

##### Гипотезы о коэффициентах корреляции

###### о равенстве коэффициентов корреляции определенному значению

H0: коэффициент корреляции между уровнем физической активности и уровнем холестерина у пациентов с сердечно-сосудистыми заболеваниями равен 0.6.

Для проверки этой гипотезы исследователи могли бы собрать данные о физической активности и уровне холестерина у группы пациентов с сердечно-сосудистыми заболеваниями. Рассчитать коэффициент корреляции между этими показателями.

Затем с помощью статистических методов можно было бы оценить, насколько полученное значение коэффициента корреляции близко к утвержденному значению 0.6. Результаты такого исследования могли бы помочь лучше понять взаимосвязь между уровнем физической активности и уровнем холестерина у пациентов с сердечно-сосудистыми заболеваниями и определить, насколько сильной является эта связь.

###### о значимости различия коэффициентов корреляции двух совокупностей

H0: существует сильная отрицательная корреляция между уровнем стресса и чувствительностью к боли. Чем выше уровень стресса, тем ниже чувствительность к боли.

Исследователи интересуются зависимостью между уровнем стресса и чувствительностью к боли у пациентов с хронической мигренью. Для проверки этой гипотезы исследователи могут измерить уровень стресса и чувствительность к боли у группы пациентов с хронической мигренью. Затем использовать коэффициент корреляции Пирсона для оценки степени линейной взаимосвязи между этими двумя переменными.

Сильная отрицательная корреляция в результатах исследования может подтвердить гипотезу о взаимосвязи между уровнем стресса и чувствительностью к боли у пациентов с мигренью. Полученные выводы могут быть полезными для разработки индивидуализированных методов лечения мигрени, учитывающих влияние стресса на чувствительность к боли у пациентов.

##### Гипотезы о долях признака

###### о равенстве доли признака определенному значению

H0: доля пациентов, достигших полной ремиссии после лечения острого лимфобластного лейкоза у детей, равна 90% при использовании протокола лечения, включающего высокодозную химиотерапию и трансплантацию стволовых клеток.

Гипотеза фокусируется на конкретном заболевании (острый лимфобластный лейкоз) и конкретной популяции пациентов (дети). Она устанавливает конкретное значение доли признака (90% достигнутой полной ремиссии) и связывает его с конкретным протоколом лечения.

Гипотеза поддается проверке путем сбора данных о результатах лечения у большой выборки пациентов.

###### о значимости различия долей признака в двух совокупностях

H0: доля пациентов с определенным генетическим вариантом, обладающих высокой чувствительностью к определенному лекарству, выше среди женщин, чем среди мужчин.

Гипотеза заключается в том, что существует различие в доле пациентов с определенным генетическим признаком, у которых выявлена высокая чувствительность к определенному лекарству, между мужчинами и женщинами. Для проверки этой гипотезы исследователи могут провести анализ генетических данных и реакции на лекарство у пациентов различного пола. Собрав эти данные у пациентов обоих полов, исследователи могут определить долю пациентов с этим признаком среди женщин и мужчин.

Далее можно сравнить полученные доли и провести статистический анализ проверки гипотезы о возможном различии между полами в отношении чувствительности к лекарству. Подтверждение гипотезы может помочь в разработке индивидуализированных схем лечения, учитывающих генетические особенности и различия между полами в реакции на определенные лекарства.

##### Гипотезы о независимости признаков в корреляционной таблице

H0: уровень физической активности пациентов с заболеванием суставов не коррелирует с уровнем воспаления в их организме.

Исследователи собирают данные о физической активности пациентов (например, количество физических нагрузок в неделю) и об уровне воспаления (например, уровень C-реактивного белка в крови). После сбора данных они создают корреляционную таблицу и проводят корреляционный анализ между уровнем физической активности и уровнем воспаления. Если же статистически значимой корреляции между этими переменными не будет выявлено, то исследователи могут принять гипотезу о независимости этих признаков.

Это может быть важной информацией для разработки индивидуализированных планов лечения и реабилитации пациентов с проблемами суставов.

##### По направленности

###### направленные
Н0: эффективность нового препарата в лечении определенного заболевания не превышает эффективность стандартного лечения.

Такая гипотеза предполагает, что новый препарат не принесет значимого клинического улучшения по сравнению с уже существующими методиками лечения. Исследование, проведенное для проверки этой гипотезы, может выявить, есть ли статистически значимое различие в эффективности нового препарата по сравнению с применяемым стандартным лечением.

###### ненаправленные

Н0: средний уровень холестерина у пациентов, принимающих новый препарат, не отличается от среднего уровня холестерина у пациентов, принимающих плацебо.

Проверка этой гипотезы подразумевает сравнение средних значений холестерина в крови у обеих групп пациентов. Чтобы выявить, есть ли между ними статистически значимое различие. Если результаты исследования не отклоняют нулевую гипотезу, то это может означать отсутствие значимых различий в уровне холестерина у пациентов, принимающих новый препарат, и тех, кто принимает плацебо.


Направленную статистическую гипотезу формулируют для проверки правдивости утверждения, что в группе А произошли более сильные изменения, чем в группе В.

Ненаправленную статистическую гипотезу формулируют, если необходимо проверить, различаются ли в разных группах формы распределения исследуемого признака.

Статистические гипотезы оцениваются с помощью расчета статистических критериев.
