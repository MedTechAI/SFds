- **Instruct Phi** — семейство компактных языковых моделей от Microsoft (Phi-3, Phi-2 и др.), оптимизированных для выполнения инструкций. Хорошо подходят для работы на ограниченных вычислительных ресурсах.

### Аппаратные ресурсы

````python
import multiprocessing

cpu_cores = psutil.cpu_count(logical=False) or multiprocessing.cpu_count()
cpu_threads = psutil.cpu_count(logical=True) or multiprocessing.cpu_count()
ram_gb = psutil.virtual_memory().total / 1e9

# Обнаружение GPU: CUDA (NVIDIA) → MPS (Apple Silicon) → CPU (fallback).
# torch.cuda.is_available() может зависнуть на Windows при первом вызове —
# весь блок обёрнут в try/except с таймаутом на уровне импорта (import torch
# выполнен ранее, CUDA-рантайм уже прогрелся). При недоступности GPU —
# автоматический переход на CPU, все переменные гарантированно определены.
try:
    gpu_available = torch.cuda.is_available()
    gpu_count = torch.cuda.device_count() if gpu_available else 0
    if gpu_available:
        device = torch.device('cuda')
        try:
            gpu_name = torch.cuda.get_device_name(0) or 'CUDA-устройство'
            gpu_vram = round(torch.cuda.get_device_properties(0).total_mem / 1e9, 1)
        except Exception:
            gpu_name = 'CUDA-устройство'
            gpu_vram = 0
    elif torch.backends.mps.is_available():
        device = torch.device('mps')
        gpu_name = 'Apple Silicon (MPS)'
        gpu_vram = 0
    else:
        device = torch.device('cpu')
        gpu_name = 'нет'
        gpu_vram = 0
except Exception:
    gpu_available = False
    gpu_count = 0
    device = torch.device('cpu')
    gpu_name = 'нет'
    gpu_vram = 0

# Производные параметры, используемые во всех ячейках ниже.
# RAM_SAFETY_FACTOR — доля RAM для данных (0.5 = 50%, оставляет резерв для ОС).
# N_JOBS — число физических ядер (core), не потоков (thread).
# BATCH_SIZE — адаптивный: 64 для GPU, вычисляется из RAM для CPU.
# DL_WORKERS — для PyTorch DataLoader: все ядра на GPU, 0 на CPU.
RAM_SAFETY_FACTOR = 0.5
N_JOBS = cpu_cores
THREADS_BOOST = -1
BATCH_SIZE_GPU = 64
BATCH_SIZE_CPU = max(4, int(ram_gb * 2))
BATCH_SIZE = BATCH_SIZE_GPU if gpu_available else BATCH_SIZE_CPU
DL_WORKERS = cpu_cores if gpu_count else 0

print(
    f'CPU: {cpu_cores} ядер / {cpu_threads} потоков | '
    f'RAM: {ram_gb:.1f} GB | '
    f'GPU: {gpu_name}'
    + (f' ({gpu_vram:.1f} GB)' if gpu_vram else '')
    + f' | device: {device}'
)
print(f'N_JOBS={N_JOBS} | BATCH_SIZE={BATCH_SIZE} | DL_WORKERS={DL_WORKERS}')
````

### Утилиты и структура папок

````python
_MODELS_DIR = Path('_models')
_ANALYSIS_DIR = Path('_analysis')
_EXAMPLES_DIR = Path('_examples')
_PLOTS_DIR = Path('_plots')
_DATA_DIR = Path('_data')
_CONFIG_DIR = Path('_config')
_TEMP_DIR = Path('_temp')
for _d in [_MODELS_DIR, _ANALYSIS_DIR, _EXAMPLES_DIR, _PLOTS_DIR, _DATA_DIR, _CONFIG_DIR, _TEMP_DIR]:
    _d.mkdir(exist_ok=True)

DISPLAY_LIMIT = 60
SUMMARY_ROWS = 50

def save_and_show(df, filename, directory=_ANALYSIS_DIR):
    """Сохраняет DataFrame в CSV и выводит в ноутбук (полностью или head).
    Файл всегда полный, без обрезания. В ячейку выводится: весь df при числе
    строк <= DISPLAY_LIMIT, иначе первые SUMMARY_ROWS строк. Используется
    для всех аналитических таблиц в {_ANALYSIS_DIR}/ и демо-вывода в {_EXAMPLES_DIR}/.
    Args:
        df (pd.DataFrame): данные для сохранения.
        filename (str): имя CSV-файла.
        directory (Path, optional): целевая папка. По умолчанию _ANALYSIS_DIR.
    Returns:
        None. Побочный эффект — файл на диске и вывод в ячейку."""
    filepath = directory / filename
    df.to_csv(filepath, index=False)
    if len(df) <= DISPLAY_LIMIT:
        display(df)
    else:
        display(df.head(SUMMARY_ROWS))
    print(f'{filename}: {len(df)} строк -> {filepath}')
def save_text(text, filename, directory=_ANALYSIS_DIR):
    """Сохраняет произвольный текст в файл без вывода в ноутбук.
    Используется для отчётов, метрик, параметров и любой текстовой аналитики.
    Файл всегда полный, без обрезания.
    Args:
        text (str): текст для сохранения.
        filename (str): имя файла.
        directory (Path, optional): целевая папка. По умолчанию _ANALYSIS_DIR.
    Returns:
        None. Побочный эффект — файл на диске."""
    filepath = directory / filename
    filepath.write_text(str(text), encoding='utf-8')
    print(f'{filename} -> {filepath}')
print('Утилиты загружены, папки созданы')
````

````python
import threading
from collections import deque
@dataclass
class HardwareMonitor:
    """Фоновый мониторинг аппаратных ресурсов: CPU%, RAM, температура.
    Запускается потоком через monitor.start(), останавливается monitor.stop().
    Каждые interval секунд пишет статус в ноутбук и логирует в resource_log.csv.
    При перегреве выше cpu_temp_critical — аварийная остановка.
    При превышении cpu_temp_cooldown — пауза на cooldown_seconds.
    Температура читается через psutil (Linux/Mac) или PowerShell WMI (Windows);
    при недоступности выводится 'n/a' — это не ошибка, а ограничение железа."""
    interval: int = field(default=30)
    cpu_temp_warn: int = field(default=85)
    cpu_temp_critical: int = field(default=95)
    cpu_temp_cooldown: int = field(default=90)
    cooldown_seconds: int = field(default=60)
    ram_warn_pct: float = field(default=0.97)
    _stop: bool = field(default=False, init=False)
    _thread: Any = field(default=None, init=False)
    _log: list = field(default_factory=list, init=False)
    _lines: deque = field(default_factory=lambda: deque(maxlen=5), init=False)
    def _get_status(self):
        """Собирает текущие показатели CPU, RAM и температуры."""
        try:
            temps = psutil.sensors_temperatures()
            max_temp = max(entry.current for name, entries in temps.items() for entry in entries if entry.current) if temps else None
        except Exception:
            max_temp = None
        if max_temp is None:
            try:
                import subprocess as _sp
                raw = _sp.check_output(
                    ['powershell', '-Command',
                     'Get-WmiObject MSAcpi_ThermalZoneTemperature -Namespace root/wmi | ForEach-Object { $_.CurrentTemperature }'],
                    timeout=3, text=True).strip()
                if raw:
                    temps = [float(r) / 10.0 - 273.15 for r in raw.split() if r.strip()]
                    max_temp = max(temps) if temps else None
            except Exception:
                pass
        ram_used_gb = psutil.virtual_memory().used / 1e9
        cpu_percent = psutil.cpu_percent(interval=0.1)
        return {'temp_max': max_temp, 'ram_used_gb': ram_used_gb, 'cpu_percent': cpu_percent}
    def _warn(self, status):
        """Выводит предупреждения при превышении порогов CPU или RAM.
        Args:
            status (dict): словарь с ключами temp_max, ram_used_gb, cpu_percent
        """
        warnings = []
        if status['temp_max'] is not None and status['temp_max'] > self.cpu_temp_critical:
            warnings.append(f"КРИТИЧЕСКИ: процессор {status['temp_max']:.1f}°C \u2014 порог {self.cpu_temp_critical}°C")
        elif status['temp_max'] is not None and status['temp_max'] > self.cpu_temp_warn:
            warnings.append(f"Процессор {status['temp_max']}°C превысил порог {self.cpu_temp_warn}°C")
        ram_limit = ram_gb * self.ram_warn_pct
        if status['ram_used_gb'] > ram_limit:
            warnings.append(f"ОЗУ {status['ram_used_gb']:.1f} из {ram_gb:.1f} ГБ")
        if warnings:
            for w in warnings:
                print(f'[Hardware warn] {w}')
        return bool(warnings)
    def status_line(self):
        """Возвращает краткую строку статуса для логирования."""
        s = self._get_status()
        result = {}
        if s['temp_max'] is not None:
            result['t°'] = f'{s["temp_max"]}°C'
        result['CPU%'] = f'{s["cpu_percent"]}%'
        result['RAM'] = f'{s["ram_used_gb"]:.1f}/{ram_gb:.1f}G'
        return result
    def _loop(self):
        """Основной цикл фонового мониторинга, запускается в отдельном потоке."""
        step = 0
        while not self._stop and threading.current_thread() is self._thread:
            time.sleep(self.interval)
            status = self._get_status()
            step += 1
            elapsed = step * self.interval
            cpu = status['cpu_percent']
            tmp = status['temp_max'] if status['temp_max'] is not None else 0.0
            ram = status['ram_used_gb']
            ram_pct = psutil.virtual_memory().percent
            line = (f'[{elapsed:4d}s]  CPU load: {cpu:4.1f}%  |  '
                    f'temp: {tmp:.1f}C  |  '
                    f'RAM: {ram_pct:4.1f}% ({ram:.1f} / {ram_gb:.1f} GB)')
            self._lines.append(line)
            if step % 6 == 0 and self._log:
                try:
                    pd.DataFrame(self._log).to_csv(_MODELS_DIR / 'resource_log.csv', index=False)
                except Exception:
                    pass
            print(self._lines[-1])
            import datetime as _dt
            _p = psutil.Process()
            _cpu_freq = psutil.cpu_freq()
            self._log.append({
                't': _dt.datetime.now(),
                'cpu': status['cpu_percent'],
                'cpu_freq': round(_cpu_freq.current, 1) if _cpu_freq else None,
                'ram_pct': psutil.virtual_memory().percent,
                'ram_process': round(_p.memory_info().rss / 1e9, 2),
                'cpu_per_core': list(psutil.cpu_percent(percpu=True)),
                'temp': status['temp_max'],
            })
            self._warn(status)
            if status['temp_max'] is not None and status['temp_max'] > self.cpu_temp_cooldown and status['temp_max'] < self.cpu_temp_critical:
                print(f'[Cooldown] temp {status["temp_max"]:.1f}°C > {self.cpu_temp_cooldown}°C \u2014 pause {self.cooldown_seconds}s')
                time.sleep(self.cooldown_seconds)
            if status['temp_max'] is not None and status['temp_max'] > self.cpu_temp_critical:
                print(f'[CRITICAL] temp {status["temp_max"]:.1f}°C > {self.cpu_temp_critical}°C \u2014 shutdown')
                import os as _os
                _os._exit(1)
    def start(self):
        """Запускает фоновый мониторинг в отдельном потоке-демоне."""
        self._stop = True
        if self._thread is not None and self._thread.is_alive():
            self._thread.join(timeout=self.interval * 2)
        import sys; sys.stdout.flush()
        self._log.clear()
        self._lines.clear()
        self.status()
        self._thread = threading.Thread(target=self._loop, daemon=True)
        self._stop = False
        self._thread.start()
    def stop(self):
        """Останавливает фоновый мониторинг."""
        self._stop = True
        if self._thread is not None:
            self._thread.join(timeout=self.interval * 2)
    def clear_log(self):
        """Очищает накопленный лог мониторинга."""
        self._log.clear()
    def get_log(self):
        """Возвращает накопленный лог мониторинга как DataFrame."""
        return pd.DataFrame(self._log) if self._log else pd.DataFrame()
    def status(self):
        """Выводит текущий статус аппаратных ресурсов."""
        s = self._get_status()
        print('Hardware status:')
        if s['temp_max'] is not None:
            print(f"  CPU temp: {s['temp_max']:.1f}C (warn: {self.cpu_temp_warn}C)")
        else:
            print('  CPU temp: n/a (ACPI zone, not core sensor)')
        print(f"  CPU load: {s['cpu_percent']}%")
        ram_pct = psutil.virtual_memory().percent
        print(f"  RAM: {ram_pct:.1f}% ({s['ram_used_gb']:.1f} / {ram_gb:.1f} GB, warn: {self.ram_warn_pct*100:.0f}%)")
        print(f"  Device: {device}")
        if device.type == 'cuda':
            print(f'  GPU: {gpu_name}, VRAM: {gpu_vram} GB')
        print(f'  Monitor: every {self.interval} sec')
monitor = HardwareMonitor(interval=10)
monitor.status()
````

````python
def info_toDF(df0):
    """Конвертирует вывод df.info() в DataFrame для сохранения в CSV.
    Извлекает имена колонок, типы, число non-null и процент заполнения.
    Используется для быстрого аудита схемы датасета при смене входных данных."""
    df = df0.copy()
    df.columns = [j.replace(' ','_') for j in df.columns]
    buffer = io.StringIO()
    df.info(buf=buffer)
    s = buffer.getvalue()
    info_df = pd.DataFrame( [line.split() for line in s.splitlines()[3:-2]] ) .fillna('') .drop(0, axis=1)
    info_df.columns = info_df.loc[0].values
    info_df = info_df.iloc[2:]
    info_df['Non-Null'] = info_df['Non-Null'].astype(int)
    info_df['Null'] = df0.shape[0] - info_df['Non-Null']
    info_df['% Non-Null'] = (info_df['Non-Null'] * 100) / df0.shape[0] 
    return info_df .set_index('Column')
def get_max_features(X_train, ngram_range=(1, 2)):
    """Вычисляет оптимальный размер TF-IDF словаря из доступной RAM.
    Формула: min(vocab_size, ram_gb * 1e6 / 100 * RAM_SAFETY_FACTOR).
    При смене железа или данных пересчитывается автоматически."""
    temp = TfidfVectorizer(max_features=None, ngram_range=ngram_range)
    vocab_size = len(temp.fit(X_train).get_feature_names_out())
    result = min(vocab_size, int(ram_gb * 1e6 / 100 * RAM_SAFETY_FACTOR))
    print(f'MAX_FEATURES = {result} (vocab={vocab_size})')
    return result
def get_batch_size():
    """Вычисляет размер батча: 64 для GPU, адаптивно из RAM для CPU.
    Формула CPU: max(16, min(128, ram_gb * RAM_SAFETY_FACTOR * 20)).
    Используется в RuBERT, MLP и DataLoader. При смене железа — авто."""
    if gpu_available:
        return 64
    return max(16, min(128, int(ram_gb * RAM_SAFETY_FACTOR * 20)))
def get_trials():
    """Вычисляет число Optuna-итераций: 50 для GPU, адаптивно из RAM для CPU.
    Формула CPU: max(10, ram_gb * RAM_SAFETY_FACTOR * 0.5).
    При смене железа пересчитывается автоматически."""
    if gpu_available:
        return 50
    return max(10, int(ram_gb * RAM_SAFETY_FACTOR * 0.5))
````

### Нормализация таргета — класс TargetNormalizer

---
<span style="font-size:0.8em;"> [TOC](#mod1) </span>


````python
try:
    nltk.download('stopwords', quiet=True)
except Exception:
    pass  # fallback при отсутствии сети

@dataclass
class TargetNormalizer:
    """Нормализует сырые названия врачей к специальностям Приказа 436н.
    Четырёхуровневая стратегия: abbrev (словарь) → prefix/substring → fuzzy →
    semantic search (SentenceTransformer). Кеширует названия в _name_tokens
    для устранения повторных вызовов pymorphy3 (O(1) lookup после init).
    Поддерживает два метода: prefix (основной, 0% low confidence) и cross
    (CrossEncoder для re-ranking, экспериментальный)."""
    official_names: List[str]
    model_name: str = field(default='DeepPavlov/rubert-base-cased')
    model: Any = field(default=None)
    embeddings: Any = field(default=None)
    stopwords: set = field(default_factory=lambda: set(nltk.corpus.stopwords.words('russian')) | {'врач', 'доктор', 'специалист', 'консультант', 'общей', 'общая', 'общий'})
    prefix_min_len: int = field(default=5)
    short_word_threshold: int = field(default=5)
    substring_score: float = field(default=0.80)
    prefix_score: float = field(default=0.85)
    match_threshold: float = field(default=0.72)
    cross_encoder_name: str = field(default='DiTy/cross-encoder-russian-msmarco')
    def _init_model(self):
        """Ленивая инициализация: SentenceTransformer + кеш токенов названий.
        Выполняется однократно при первом вызове любого метода матчинга.
        _name_tokens — предразобранные токены всех ~113 специальностей,
        устраняет 300+ повторных вызовов pymorphy3 на каждое слово."""
        if self.model is None:
            self.model = SentenceTransformer(self.model_name)
            self.embeddings = self.model.encode(self.official_names, convert_to_tensor=True)
            self._morph = pymorphy3.MorphAnalyzer()
            self._cross_encoder = CrossEncoder(self.cross_encoder_name)
            # Уровень 0: словарь медицинских аббревиатур (жёсткое сопоставление).
            # Тактические соображения: короткие аббревиатуры (ЛОР, ЖКТ, ЛФК...)
            # семантический эмбеддинг не различает из-за малой длины. Альтернативы:
            # - внешний конфиг {_CONFIG_DIR}/medical_abbreviations.json (расширяемо);
            # - частотный анализ low-confidence слов + ручная экспертная разметка;
            # - отдельный проект по сбору справочника МКБ-10 аббревиатур.
            # Текущий подход — компромисс: словарь в коде, при росте объёма —
            # вынос в {_CONFIG_DIR}/.
            self._abbrev = {
                            'лор': 'Оториноларингология',
                            'жкт': 'Гастроэнтерология',
                'лфк': 'Лечебная физкультура и спортивная медицина',
                'воп': 'Общая врачебная практика (семейная медицина)',
                'терпевт': 'Терапия',
                'репродуктолог': 'Акушерство и гинекология',
                'маммолог': 'Онкология',
                'мамолог': 'Онкология',
                'гепатолог': 'Гастроэнтерология',
                'андролог': 'Урология',
                'нейролог': 'Неврология',
                'дерматокосметолог': 'Косметология',
                'астмолог': 'Пульмонология',
                'семейная': 'Общая врачебная практика (семейная медицина)',
                'семейный': 'Общая врачебная практика (семейная медицина)',
                'plasticology': 'Пластическая хирургия',
                        }
            self._name_tokens = []
            for name in self.official_names:
                tokens = re.split(r'[\s-]+', name.lower())
                self._name_tokens.append((name, tokens))
    def _clean_target(self, value: str) -> List[str]:
        self._init_model()
        raw_words = re.findall(r'[а-яёa-z0-9-]+', value)
        words = []
        for w in raw_words:
            if w in self.stopwords:
                continue
            words.extend(part for part in w.split('-') if part and part not in self.stopwords)
        return words if words else [value]
    def _match_fuzzy(self, word: str):
        """Уровень 2: fuzzy matching через SequenceMatcher для опечаток.
        Сравнивает слово с каждым токеном каждого названия специальности
        из предварительно закешированного _name_tokens. Возвращает название
        с максимальным ratio (0..1). Исправляет: 'терпевт' → 'Терапия'."""
        best_name, best_score = None, 0.0
        for name, name_tokens in self._name_tokens:
            for token in name_tokens:
                dist = SequenceMatcher(None, word, token).ratio()
                if dist > best_score:
                    best_score, best_name = dist, name
        return best_name, best_score
    def _match_cross(self, word: str):
        """Экспериментальный метод: semantic search → топ-3 → CrossEncoder.
        SentenceTransformer быстро находит 3 кандидата по косинусному сходству,
        затем CrossEncoder точно переранжирует пары (word, candidate).
        Медленнее prefix, но потенциально точнее для сложных случаев."""
        self._init_model()
        word_clean = self._morph.parse(word)[0].normal_form
        emb = self.model.encode(word_clean, convert_to_tensor=True)
        candidates = util.semantic_search(emb, self.embeddings, top_k=3)[0]
        top3_names = [self.official_names[c['corpus_id']] for c in candidates]
        pairs = [(word_clean, name) for name in top3_names]
        scores = self._cross_encoder.predict(pairs, batch_size=max(4, cpu_threads*2))
        best_idx = int(scores.argmax())
        return top3_names[best_idx], float(scores[best_idx])
    def _match_prefix(self, word: str):
        """Основной метод нормализации: prefix + fuzzy + semantic fallback.
        Уровень 1: строковое совпадение — префикс токена или вхождение
        короткого слова в название специальности. Score = substring_score или
        prefix_score. При нескольких кандидатах — tie-break по длине общего
        префикса (commonprefix). Порог: match_threshold (0.72).
        Уровень 2: fuzzy matching через SequenceMatcher (опечатки).
        Уровень 3: semantic search через SentenceTransformer (последний шанс)."""
        self._init_model()
        word_clean = self._morph.parse(word)[0].normal_form
        # Уровень 0: жёсткий словарь медицинских аббревиатур
        if word_clean in self._abbrev:
            return self._abbrev[word_clean], 1.0
        prefix = word_clean[:max(len(word_clean)//2, self.prefix_min_len)]
        best, best_ratio = None, 0.0
        for name, name_tokens in self._name_tokens:
            for token in name_tokens:
                if len(word_clean) < self.short_word_threshold:
                    if word_clean in token:
                        sim = self.substring_score
                    else:
                        continue
                elif token.startswith(prefix):
                    sim = self.prefix_score
                else:
                    continue
                if sim > best_ratio or (sim == best_ratio and best is not None and len(os.path.commonprefix([word_clean, token])) > len(os.path.commonprefix([word_clean, best.lower()]))):
                    best_ratio, best = sim, name
        if best_ratio > self.match_threshold:
            return best, best_ratio
        fuzzy_best, fuzzy_score = self._match_fuzzy(word_clean)
        if fuzzy_score > 0.85:
            return fuzzy_best, fuzzy_score
        emb = self.model.encode(word_clean, convert_to_tensor=True)
        m = util.semantic_search(emb, self.embeddings, top_k=1)[0][0]
        return self.official_names[m['corpus_id']], m['score']
        
        
    def fit_transform(self, df: pd.DataFrame, source_col: str = 'to_doctor', method: str = 'prefix') -> pd.DataFrame:
        """Применяет нормализацию к датасету: одна строка в несколько.
        Args:
            df (pd.DataFrame): исходный датасет
            source_col (str): колонка с сырыми названиями врачей
            method (str): метод матчинга (prefix/cross)
        Returns:
            pd.DataFrame: датасет с нормализованным таргетом
        """
        raw = df[source_col].str.strip().str.lower()
        word_lists = raw.apply(self._clean_target)
        expanded = []
        for _, row in tqdm(df.iterrows(), total=len(df), desc='TargetNormalizer'):
            words = word_lists.loc[row.name]
            for w in words:
                new_row = row.to_dict()
                new_row['to_doctor_clean'] = w
                if method == 'cross':
                    new_row['target_normalized'], new_row['match_score'] = self._match_cross(w)
                else:
                    new_row['target_normalized'], new_row['match_score'] = self._match_prefix(w)
                expanded.append(new_row)
        result = pd.DataFrame(expanded).reset_index(drop=True)
        low_mask = result['match_score'] < 0.5
        print(f'Исходных строк: {len(df)}')
        assert len(result) >= len(df), 'data loss'
        print(f'После нормализации: {len(result)} строк')
        print(f'Уникальных классов: {result["target_normalized"].nunique()}')
        print(f'Low confidence (score < 0.5): {low_mask.sum()} ({(low_mask.sum()/len(result)*100):.1f}%)')
        if low_mask.sum() > 0:
            print('\nЗначения с низкой уверенностью (сохранены):')
            low_df = result[low_mask][['to_doctor_clean', 'target_normalized', 'match_score']].drop_duplicates()
            for _, row in low_df.iterrows():
                print(f'  {row["to_doctor_clean"]:30s} -> {row["target_normalized"]:45s} (score: {row["match_score"]:.3f})')
        return result

normalizer = TargetNormalizer(official_names=OFFICIAL_NAMES_436N)
pass  # _ANALYSIS_DIR created globally
COLS_OUT = ['to_doctor_clean', 'target_normalized', 'match_score']

def save_results(df, name):
    """Сохраняет результаты матчинга в CSV и возвращает подмножество.
    Args:
        df (pd.DataFrame): данные после нормализации
        name (str): имя метода (prefix/cross_encoder)
    Returns:
        уникальные пары слово-специальность, сортированные по score
        """
    subset = df[COLS_OUT].drop_duplicates().sort_values('match_score')
    save_and_show(subset, f'match_{name}.csv')
    return subset

print('=== Prefix-based матчинг ===')
monitor.start()
df_multirow = normalizer.fit_transform(df, method='prefix')
monitor.stop()
prefix_subset = save_results(df_multirow, 'prefix')
print(f'{len(prefix_subset)} уникальных пар -> match_prefix.{{csv,txt}}')

print('\n=== Cross-encoder матчинг ===')
df_cross = normalizer.fit_transform(df, method='cross')
cross_subset = save_results(df_cross, 'cross_encoder')
print(f'{len(cross_subset)} уникальных пар -> match_cross_encoder.{{csv,txt}}')

merged = (df_multirow[['user_question'] + COLS_OUT]
          .merge(df_cross[['user_question'] + COLS_OUT],
                 on=['user_question', 'to_doctor_clean'],
                 suffixes=('_prefix', '_cross')))
diff_mask = merged['target_normalized_prefix'] != merged['target_normalized_cross']
n_total, n_diff = len(merged), diff_mask.sum()
print(f'\n=== Сравнение методов ===')
print(f'Совпадений: {n_total - n_diff} ({(1 - n_diff/n_total)*100:.1f}%)')
print(f'Расхождений: {n_diff} ({n_diff/n_total*100:.1f}%)')

report = (
    f'Сравнение методов матчинга\n'
    f'Всего пар: {n_total}\n'
    f'Совпадений: {n_total - n_diff} ({(1 - n_diff/n_total)*100:.1f}%)\n'
    f'Расхождений: {n_diff} ({n_diff/n_total*100:.1f}%)\n\n'
)
if n_diff:
    diff_df = merged[diff_mask][['to_doctor_clean', 'target_normalized_prefix', 'target_normalized_cross', 'match_score_prefix', 'match_score_cross']]
    diff_df.columns = ['Слово', 'Prefix', 'Cross-encoder', 'Score P', 'Score C']
    display(diff_df.sort_values('Слово'))
    report += diff_df.to_string(index=False)
save_text(report, 'match_comparison.txt')
print(f'''
ВЫВОД:
TargetNormalizer - четырехуровневая нормализация таргета:
уровень 0 - словарь медицинских аббревиатур (жесткое сопоставление),
уровень 1 - строковое совпадение по префиксу/подстроке,
уровень 2 - fuzzy-матчинг для опечаток,
уровень 3 - семантический поиск (последний fallback).
Исходно {len(df)} строк, после нормализации - {len(df_multirow)} строк
(мульти-специальное разбиение). Уникальных классов: {df_multirow["target_normalized"].nunique()}.
Результаты: {_ANALYSIS_DIR}/match_prefix.csv, cross_encoder_comparison.
''')
````


### Эксперимент: сравнение методов очистки слов перед semantic_search

---
<span style="font-size:0.8em;"> [TOC](#mod1) </span>


````python
FALLBACK_SCORE_THRESHOLD = 0.75


MEDICAL_SUFFIXES = {'лог', 'логия', 'ист', 'атр',
                    'ия', 'оз', 'ит', 'ома',
                    'ектомия', 'графия',
                    'скопия', 'патия'}

SNOWBALL_RU = SnowballStemmer('russian')

def strip_root(word):
    """Отрезает медицинские суффиксы от стеммированного слова.
    Args:
        word (str): исходное слово, название врача
    Returns:
        str: корень слова без медицинского суффикса"""
    s = SNOWBALL_RU.stem(word)
    for suf in sorted(MEDICAL_SUFFIXES, key=len, reverse=True):
        if s.endswith(suf) and len(s) - len(suf) >= 3:
            return s[:-len(suf)]
    return s

mapping_fallback = df_multirow[COLS_OUT].drop_duplicates()
low_conf_words = sorted(set(
    mapping_fallback[mapping_fallback['match_score'] < FALLBACK_SCORE_THRESHOLD]['to_doctor_clean']))

print(f'Слов с match_score < FALLBACK_SCORE_THRESHOLD: {len(low_conf_words)}')

comparison = []
for w in low_conf_words:
    lemma = normalizer._morph.parse(w)[0].normal_form
    snow = SNOWBALL_RU.stem(w)
    stripped = strip_root(w)
    comparison.append({'word': w, 'pymorphy3': lemma, 'Snowball': snow, 'SUFFIXES': stripped})

df_strip = pd.DataFrame(comparison)
display(df_strip)
save_and_show(df_strip, 'stripping_comparison.csv')
print(f'Таблица сохранена: {_MODELS_DIR}/stripping_comparison.csv')

words_changed = sum(
    1 for _, r in df_strip.iterrows()
    if r['SUFFIXES'] != r['pymorphy3'])

print(f'Стриппинг изменил корень у {words_changed}/{len(df_strip)} слов')

for _, r in df_strip.iterrows():
    if r['SUFFIXES'] != r['pymorphy3']:
        print(f'  {r["word"]:20s} -> {r["SUFFIXES"]}')
print(f'''
ВЫВОД:
Сравнение трех методов приведения слов к основе: лемматизация,
стемминг и отрезание медицинских суффиксов. Лемматизация через
pymorphy3 дает наиболее предсказуемый результат для TF-IDF векторизации.
Таблица: {_ANALYSIS_DIR}/stripping_comparison.csv.
''')
````


## Инструментарий предобработки текста

---
<span style="font-size:0.8em;"> [TOC](#mod1) </span>


````python
@dataclass
class TextPreprocessor:
    """Очистка и токенизация текста жалоб: lower + regex + стоп-слова.
    clean_text — удаляет спецсимволы, приводит к нижнему регистру.
    tokenize — разбивает на слова, опционально фильтрует стоп-слова NLTK."""
    stopwords: set = field(default_factory=lambda: set(nltk.corpus.stopwords.words('russian')))
    def clean_text(self, text: str) -> str:
        text = text.lower()
        text = re.sub(r'[^а-яёa-z0-9\s]', ' ', text)
        text = re.sub(r'\s+', ' ', text).strip()
        return text
    def tokenize(self, text: str, remove_stopwords: bool = True, min_len: int = 2) -> List[str]:
        words = re.findall(r'[а-яёa-z0-9]+', text.lower())
        if remove_stopwords:
            words = [w for w in words if w not in self.stopwords and len(w) > min_len]
        return words
@dataclass
class TextAnalyzer:
    """Частотный анализ: слова, n-граммы, облако слов.
    get_word_frequencies — счётчик слов после токенизации.
    get_ngrams — sliding window для биграмм/триграмм.
    generate_wordcloud — визуализация через matplotlib.wordcloud."""
    preprocessor: TextPreprocessor = field(default_factory=TextPreprocessor)
    def get_word_frequencies(self, texts: List[str]) -> Counter:
        """Подсчитывает частоты слов после токенизации.
        Args:
            texts (List[str]): список текстов жалоб
        Returns:
            Counter: счётчик частот слов
        """
        all_words = []
        for text in texts:
            words = self.preprocessor.tokenize(text)
            all_words.extend(words)
        return Counter(all_words)
    def get_ngrams(self, texts: List[str], n: int = 2) -> Counter:
        """Строит n-граммы скользящим окном.
        Args:
            texts (List[str]): список текстов
            n (int): размер n-граммы (2=биграммы, 3=триграммы)
        Returns:
            Counter: счётчик n-грамм
        """
        all_ngrams = []
        for text in texts:
            words = self.preprocessor.tokenize(text)
            if len(words) >= n:
                all_ngrams.extend(zip(*[words[i:] for i in range(n)]))  # sliding window: n shifted copies zipped
        return Counter(all_ngrams)
    def generate_wordcloud(self, frequencies: Counter, width: int = 800, height: int = 400):
        """Генерирует облако слов из частотного словаря.
        Args:
            frequencies (Counter): частоты слов
            width (int): ширина изображения
            height (int): высота изображения
        Returns:
            WordCloud: объект облака слов
        """
        return WordCloud(width=width, height=height, background_color='white',
                         colormap='viridis', max_words=100).generate_from_frequencies(frequencies)
preprocessor = TextPreprocessor()
analyzer = TextAnalyzer(preprocessor=preprocessor)
print(f'Загружено стоп-слов NLTK (русский): {len(preprocessor.stopwords)}')
print(f'''
ВЫВОД:
TextPreprocessor - очистка текста и токенизация: удаление спецсимволов,
нижний регистр, фильтрация стоп-слов. TextAnalyzer - частотный анализ
и n-граммы. Загружено {len(preprocessor.stopwords)} стоп-слов NLTK
(библиотека обработки естественного языка).
''')
````


## Частотный анализ слов

---
<span style="font-size:0.8em;"> [TOC](#mod1) </span>


````python
TOP_N_WORDS = 30
word_freq = analyzer.get_word_frequencies(df_multirow['user_question'].tolist())
print(f'Всего жалоб: {len(df_multirow)}')
print(f'Всего слов после очистки: {sum(word_freq.values())}')
print(f'Уникальных слов: {len(word_freq)}')
print(f'\nТоп-{TOP_N_WORDS} частотных слов:')
top_words = word_freq.most_common(TOP_N_WORDS)
for word, count in top_words:
    print(f'  {word}: {count}')
fig = px.bar(x=[w for w, _ in top_words[::-1]], y=[c for _, c in top_words[::-1]],
             orientation='h', title=f'Топ-{TOP_N_WORDS} частотных слов',
             labels={'x': 'Частота', 'y': 'Слово'},
             color=[c for _, c in top_words[::-1]], color_continuous_scale='Viridis')
fig.update_layout(height=800, showlegend=False,
    margin=dict(l=180, r=20, t=50, b=20),
    title_font=dict(size=16),
    xaxis=dict(tickfont=dict(size=11)), yaxis=dict(tickfont=dict(size=12)))
fig.show()
fig.write_html(str(_PLOTS_DIR / 'word_frequency_top30.html'))
word_freq_df = pd.DataFrame(word_freq.most_common(), columns=['word', 'count'])
save_and_show(word_freq_df, 'word_frequencies_full.csv')

#
# Топ-30 частотных слов: «боль», «мочь», «неделя», «болеть», «голова» —
# общеупотребительная медицинская лексика. Полный частотный словарь (~5000
# уникальных слов) сохранён в word_frequencies_full.csv. Частоты убывают
# по степенному закону — большинство слов встречаются редко.
print(f'''
ВЫВОД:
Частотный анализ: топ-{TOP_N_WORDS} слов. Всего {sum(word_freq.values())}
токенов, {len(word_freq)} уникальных. Частоты убывают по степенному
закону - большинство слов встречаются редко.
Полный словарь: {_ANALYSIS_DIR}/word_frequencies_full.csv.
''')
````

````python
TOP_N_COVERAGE_1 = 10
TOP_N_COVERAGE_2 = 100
freqs = np.array(sorted(word_freq.values(), reverse=True))
ranks = np.arange(1, len(freqs) + 1)
# Закон Ципфа: линейная регрессия log(ранг) ~ log(частота).
# Параметр a (|slope|) — наклон прямой, характеризующий скорость убывания частот.
# R2 — коэффициент детерминации, насколько хорошо данные описываются степенным законом.
slope, intercept, r_value, _, _ = scipy.stats.linregress(np.log(ranks), np.log(freqs))
# Кумулятивная доля: какую часть всех токенов покрывают первые N слов.
cumulative = np.cumsum(freqs) / np.sum(freqs)
fig = make_subplots(rows=1, cols=2, subplot_titles=('Закон Ципфа: ранг vs частота (log-log)', 'Кумулятивное покрытие'))
fig.add_trace(go.Scatter(x=ranks, y=freqs, mode='markers', marker_size=3, opacity=0.4, name='Слова'), row=1, col=1)
fig.add_trace(go.Scatter(x=ranks, y=np.exp(intercept) * ranks ** slope, mode='lines',
    line_color='red', name=f'Ципф: a={abs(slope):.2f} (R2={r_value**2:.3f})'), row=1, col=1)
fig.update_xaxes(type='log', title='Ранг (log)', row=1, col=1)
fig.update_yaxes(type='log', title='Частота (log)', row=1, col=1)
fig.add_trace(go.Scatter(x=ranks, y=cumulative, mode='lines', fill='tozeroy',
    line_color='steelblue'), row=1, col=2)
fig.add_vline(x=TOP_N_COVERAGE_1, line_dash='dash', line_color='red', opacity=0.5,
    annotation_text=f'Top-{TOP_N_COVERAGE_1}: {cumulative[TOP_N_COVERAGE_1-1]:.1%}', row=1, col=2)
fig.add_vline(x=TOP_N_COVERAGE_2, line_dash='dash', line_color='orange', opacity=0.5,
    annotation_text=f'Top-{TOP_N_COVERAGE_2}: {cumulative[TOP_N_COVERAGE_2-1]:.1%}', row=1, col=2)
fig.update_xaxes(title='Ранг слова', row=1, col=2)
fig.update_yaxes(title='Кумулятивная доля', row=1, col=2)
fig.update_layout(height=400, showlegend=True)
fig.show()
fig.write_html(str(_PLOTS_DIR / 'zipf_coverage.html'))
print(f'Параметр a закона Ципфа: {abs(slope):.2f} (R2={r_value**2:.3f})')
print(f'Top-{TOP_N_COVERAGE_1} слов покрывают {cumulative[TOP_N_COVERAGE_1-1]:.1%} токенов')
print(f'Top-{TOP_N_COVERAGE_2} слов покрывают {cumulative[TOP_N_COVERAGE_2-1]:.1%} токенов')
save_text(f'Zipf params: a={abs(slope):.2f}, R2={r_value**2:.3f}\nTop-{TOP_N_COVERAGE_1} coverage: {cumulative[TOP_N_COVERAGE_1-1]:.1%}\nTop-{TOP_N_COVERAGE_2} coverage: {cumulative[TOP_N_COVERAGE_2-1]:.1%}', 'zipf_params.txt')

#
print(f'''
ВЫВОД:
Закон Ципфа подтверждается: параметр a = {abs(slope):.2f},
R2 = {r_value**2:.3f}. Топ-{TOP_N_COVERAGE_1} слов покрывают
{cumulative[TOP_N_COVERAGE_1-1]:.1%} токенов, топ-{TOP_N_COVERAGE_2} -
{cumulative[TOP_N_COVERAGE_2-1]:.1%}. Словарь сконцентрирован:
небольшая группа слов используется очень часто.
Параметры: {_ANALYSIS_DIR}/zipf_params.txt, график: {_PLOTS_DIR}/zipf_coverage.html.
''')
````


## Кластеризация эмбеддингов → профили

````python
names = df_multirow['target_normalized'].unique()
normalizer._init_model()
name_emb = normalizer.model.encode(names, convert_to_tensor=True).cpu().numpy()
# Иерархическая кластеризация: эмбеддинги 34 названий специальностей.
# linkage + average — не требует евклидовой метрики (Ward: метод Уорда минимизирует внутрикластерную дисперсию, только для евклидовой) внутри кластеров при объединении.
Z = linkage(name_emb, method='average')
plt.figure(figsize=(16, 6))
dendrogram(Z, labels=names, leaf_rotation=90, leaf_font_size=7)
plt.title('Дендрограмма: кластеризация специальностей 436н')
plt.xlabel('Специальность')
plt.ylabel('Расстояние')
plt.tight_layout()
plt.savefig(str(_PLOTS_DIR / 'dendrogram_specialties.png'), dpi=100, bbox_inches='tight')
# fcluster — плоская нарезка дендрограммы: t=5 → максимум 5 профилей.
clusters = fcluster(Z, t=5, criterion='maxclust')
print(f'Получено профилей: {len(set(clusters))}')
print(f'''
ВЫВОД:
Иерархическая кластеризация специальностей через эмбеддинги
названий. Дендрограмма показывает естественные группы:
хирургические, терапевтические, детские. Разрезание на профили
используется для двухуровневой модели. График: {_PLOTS_DIR}/dendrogram_specialties.png.
''')
````

````python
sil = silhouette_score(name_emb, clusters)
ch = calinski_harabasz_score(name_emb, clusters)  # отношение межкластерной дисперсии к внутрикластерной (выше = лучше)
print(f'Качество кластеризации ({len(set(clusters))} профилей):')
print(f'  Silhouette score: {sil:.3f} (от -1 до +1, выше = лучше)')
print(f'  Calinski-Harabasz index: {ch:.1f} (выше = лучше)')
sil_values = silhouette_samples(name_emb, clusters)
plt.figure(figsize=(10, 5))
y_lower = 10
for i in sorted(set(clusters)):
    vals = sil_values[clusters == i]
    vals.sort()
    plt.fill_betweenx(np.arange(y_lower, y_lower + len(vals)), 0, vals,
                      alpha=0.7, label=f'Профиль {i}')
    y_lower += len(vals) + 5
plt.axvline(sil, color='red', linestyle='--', label=f'Средний: {sil:.3f}')
plt.xlabel('Силуэт'); plt.ylabel('Специальность')
plt.title('Силуэтный анализ кластеризации специальностей 436н')
plt.legend(loc='lower right', ncol=3, fontsize=7)
plt.tight_layout(); plt.show()
print(f'''
ВЫВОД:
Оценка качества кластеризации: силуэт (Silhouette score,
от -1 до +1 - чем выше, тем лучше разделены кластеры) = {sil:.3f}.
Индекс Calinski-Harabasz (отношение межкластерной дисперсии
к внутрикластерной) = {ch:.1f}. {len(set(clusters))} профилей
для двухуровневой модели.
''')
````

````python
BACKTRANSLATE_MAX_CHARS = 1500  # ограничение GoogleTranslator API

AUGMENTATION_THRESHOLD = 10
vc_train = pd.Series(y_train).value_counts()
rare = vc_train[vc_train < AUGMENTATION_THRESHOLD].index
monitor.start()
_t_start = time.time()
translator = GoogleTranslator(source="ru", target="en")
back_translator = GoogleTranslator(source="en", target="ru")

def _augment_text(text):
    """Back-translation: русский → английский → русский.
    Создаёт синтетический парафраз исходной жалобы с сохранением смысла.
    При ошибке сети возвращает None — текст-сирота не попадёт в датасет."""
    try:
        en = translator.translate(text[:BACKTRANSLATE_MAX_CHARS])
        return back_translator.translate(en)
    except Exception:
        return None

print(f'Классов с <{AUGMENTATION_THRESHOLD} примерами в train: {len(rare)}')
X_aug, y_aug = list(X_train.copy()), list(y_train.copy())
train_orig_idx = X_train.index
for cls in rare:
    mask = y_train == cls
    texts = X_train[mask].tolist()
    print(f'\n{le.inverse_transform([cls])[0]}: оригинальных примеров: {len(texts)}')
    with ThreadPoolExecutor(max_workers=cpu_threads) as executor:
        futures = {executor.submit(_augment_text, t): t for t in texts}
        for f in as_completed(futures):
            result = f.result()
            if result:
                X_aug.append(result); y_aug.append(cls)
    joblib.dump((X_aug, y_aug), _TEMP_DIR / 'aug_checkpoint.pkl')

n_augmented = len(X_aug) - len(train_orig_idx)
X_train = pd.Series(X_aug, index=list(train_orig_idx) + list(range(len(df_multirow), len(df_multirow) + n_augmented)))
y_train = np.array(y_aug)
_t_end = time.time()
print(f'  Время аугментации: {_t_end - _t_start:.1f} сек')
_aug_time = _t_end - _t_start
monitor.stop()
aug_stats = pd.DataFrame({'examples_before': vc_train.values, 'examples_after': pd.Series(y_train).value_counts().reindex(vc_train.index, fill_value=0).values}, index=vc_train.index)
aug_stats.index.name = 'class'
save_and_show(aug_stats, 'augmentation_stats.csv')
print(f'\nПосле аугментации: {len(X_train)} примеров (было {len(y_enc)})')

#
# Аугментация данных: back-translation (русский → английский → русский)
# через GoogleTranslator. Порог — классы с менее чем 10 примерами.
# Каждый текст переводится туда-обратно, создавая синтаксический парафраз
# с сохранением медицинского смысла. Результат: X_train дополнен
# синтетическими примерами, y_train обновлён. Статистика до/после
# сохраняется в augmentation_stats.csv. Чекпоинт каждые класс —
# aug_checkpoint.pkl в _temp/ для защиты от падения сети.
print(f'''
ВЫВОД:
Аугментация через обратный перевод (русский -> английский -> русский):
классы с менее чем {AUGMENTATION_THRESHOLD} примерами дополнены синтетическими
парафразами. Исходно в train: {len(train_orig_idx)}, после аугментации: {len(X_train)}.
Чекпоинт каждого класса сохраняется. Статистика: {_ANALYSIS_DIR}/augmentation_stats.csv.
''')
````

### Абляция: предобработка текста и n-граммы

---
<span style="font-size:0.8em;"> [TOC](#mod1) </span>


````python
stemmer_abl = SnowballStemmer('russian')

def preprocess_raw(text):
    """Токенизация без лемматизации и стемминга.
    Args:
        text (str): исходный текст жалобы
    Returns:
        str: токенизированный текст, разделённый пробелами"""
    return ' '.join(preprocessor.tokenize(text))

def preprocess_lemma(text):
    """Очистка, лемматизация через pymorphy3 и токенизация.
    Args:
        text (str): исходный текст
    Returns:
        str: лемматизированный токенизированный текст"""
    lemm = ' '.join(morph.parse(w)[0].normal_form for w in preprocessor.clean_text(text).split())
    return ' '.join(preprocessor.tokenize(lemm))

def preprocess_stem(text):
    """Стемминг через SnowballStemmer и токенизация.
    Args:
        text (str): исходный текст
    Returns:
        str: стеммированный токенизированный текст"""
    words = preprocessor.tokenize(text, remove_stopwords=False)
    stemmed = [stemmer_abl.stem(w) for w in words]
    return ' '.join([w for w in stemmed if len(w) > 1 and w not in preprocessor.stopwords])

PREPROCESSING = {
    'raw': preprocess_raw,
    'lemmatized': preprocess_lemma,
    'stemmed': preprocess_stem,
}

NGRAM_VARIANTS = [(1, 1), (1, 2), (1, 3), (1, 4), (1, 5), (1, 6)]

abl_results = []
monitor.start()
for pp_name, pp_func in PREPROCESSING.items():
    X_pp = df_multirow['user_question'].apply(pp_func)
    y_pp = df_multirow['target_normalized']
    le_abl = LabelEncoder()
    y_enc_abl = le_abl.fit_transform(y_pp)
    Xt, Xv, yt, yv = train_test_split(X_pp, y_enc_abl, test_size=0.2, random_state=42)
    
    for ng in NGRAM_VARIANTS:
        pipe = Pipeline([
            ('tfidf', TfidfVectorizer(ngram_range=ng, max_features=MAX_FEATURES)),
            ('clf', RidgeClassifier(class_weight='balanced'))
        ])
        grid = GridSearchCV(pipe, {'clf__alpha': [0.1, 1.0, 10]}, cv=3, scoring='f1_weighted', n_jobs=cpu_cores)
        grid.fit(Xt, yt)
        yp = grid.predict(Xv)
        f1 = f1_score(yv, yp, average='weighted', zero_division=0)
        acc = accuracy_score(yv, yp)
        abl_results.append({
            'preprocessing': pp_name,
            'ngram': f'{ng}',
            'ngram_range': ng,
            'f1_weighted': round(f1, 4),
            'accuracy': round(acc, 4),
            })
        print(f'  {pp_name:12s} ngram={ng}: F1={f1:.4f}, Acc={acc:.4f}')

monitor.stop()
df_abl = pd.DataFrame(abl_results).sort_values('f1_weighted', ascending=False)
save_and_show(df_abl, 'ablation_preprocessing.csv')

best = df_abl.iloc[0]
print(f'\n=== Победитель ===')
print(f'  Предобработка: {best["preprocessing"]}')
print(f'  n-граммы:      {best["ngram"]}')
print(f'  F1-weighted:   {best["f1_weighted"]:.4f}')
print(f'  Accuracy:      {best["accuracy"]:.4f}')

BEST_PREPROCESSING = best['preprocessing']
BEST_NGRAM = best['ngram_range']

fig = px.bar(df_abl, x='preprocessing', y='f1_weighted', color='ngram', barmode='group',
             title='Ablation: preprocessing vs n-grams (Ridge F1-weighted)',
             labels={'f1_weighted': 'F1-weighted', 'preprocessing': 'Предобработка', 'ngram': 'n-grams'})
fig.show()
fig.write_html(str(_PLOTS_DIR / 'ablation_preprocessing.html'))
print('График сохранён: {_PLOTS_DIR}/ablation_preprocessing.html')

print(f'\nРекомендация: предобработка={BEST_PREPROCESSING}, ngram={BEST_NGRAM}')
print('X_train/MAX_FEATURES сохранены без изменений. Для применения победителя — закомментируйте эту строку и раскомментируйте блок авто-применения ниже.')

(_CONFIG_DIR / 'ablation_preprocessing.json').write_text(
    json.dumps({'best_preprocessing': BEST_PREPROCESSING, 'best_ngram': BEST_NGRAM, 'results': abl_results}, ensure_ascii=False, indent=2),
    encoding='utf-8')
print(f'''
ВЫВОД:
Абляция предобработки: {len(PREPROCESSING)} метода очистки x
{len(NGRAM_VARIANTS)} диапазонов словосочетаний = {len(PREPROCESSING)*len(NGRAM_VARIANTS)} комбинаций.
Победитель определен по наибольшему F1-weighted (среднее гармоническое
точности и полноты с весами классов). Результаты: {_ANALYSIS_DIR}/ablation_preprocessing.csv.
Победитель не применен автоматически - требует ручной активации.
''')
````


### MLP PyTorch

<span style="font-size:0.8em;"> [TOC](#mod1) </span>


````python
class MLP(nn.Module):
    """Многослойный перцептрон: 3 скрытых слоя с BatchNorm + Dropout.
    Архитектура: Linear → BN → ReLU → Dropout (×3) → Linear.
    BatchNorm стабилизирует обучение, Dropout — регуляризация."""
    def __init__(self, n_features, n_classes, hidden1=512, hidden2=256, hidden3=128, dropout=0.3):
        super().__init__()
        self.net = nn.Sequential(
            # Linear — полносвязный слой, BatchNorm1d — нормализация активаций по батчу.
            nn.Linear(n_features, hidden1),
            nn.BatchNorm1d(hidden1),
            # ReLU — функция активации (отсекает отрицательные значения).
            nn.ReLU(),
            # Dropout — случайно обнуляет нейроны с вероятностью dropout (регуляризация).
            nn.Dropout(dropout),
            nn.Linear(hidden1, hidden2),
            nn.BatchNorm1d(hidden2),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden2, hidden3),
            nn.BatchNorm1d(hidden3),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden3, n_classes),
        )
    def forward(self, x):
        """Прямой проход через сеть.
        Args:
            x (torch.Tensor): входной батч
        Returns:
            torch.Tensor: логиты классов
        """
        return self.net(x)

N_TRIALS_MLP = get_trials()

@dataclass
class MLPParams:
    """Гиперпараметры MLP: архитектура (hidden), регуляризация, оптимизатор.
    Значения — лучшие из Optuna (Trial 2, 10 trials, F1=0.749)."""
    hidden1: int = 256
    hidden2: int = 128
    hidden3: int = 128
    dropout: float = 0.2
    lr: float = 0.00486
    weight_decay: float = 0.001104

# === Mode: uncomment = fast-path, comment = tuning ===
mlp_cfg = MLPParams()

vec_mlp = TfidfVectorizer(ngram_range=(1, 5), max_features=MAX_FEATURES).fit(X_train)
_t0 = time.time()
X_train_mlp = torch.FloatTensor(vec_mlp.transform(X_train).toarray())
X_val_mlp = torch.FloatTensor(vec_mlp.transform(X_val).toarray())
y_train_t = torch.LongTensor(y_train)
y_val_t = torch.LongTensor(y_val)
n_features = X_train_mlp.shape[1]
n_classes = len(le.classes_)

print(f'Optuna: подбор архитектуры MLP ({n_features} признаков, {N_TRIALS_MLP} trials)....')

try:
    mlp_params = mlp_cfg
    model_mlp = MLP(n_features=n_features, n_classes=n_classes, hidden1=mlp_params.hidden1, hidden2=mlp_params.hidden2, hidden3=mlp_params.hidden3, dropout=mlp_params.dropout).to(device)
    print(f'MLP (fast: {mlp_params.hidden1}/{mlp_params.hidden2}/{mlp_params.hidden3}, drop={mlp_params.dropout}, без Optuna)')
    _lr, _wd = mlp_params.lr, mlp_params.weight_decay
    best_mlp_params = {'hidden1': mlp_params.hidden1, 'hidden2': mlp_params.hidden2, 'hidden3': mlp_params.hidden3, 'dropout': mlp_params.dropout, 'lr': mlp_params.lr, 'weight_decay': mlp_params.weight_decay}
except NameError:
    # Optuna objective: байесовский поиск гиперпараметров.
    # suggest_int/suggest_float — выбор значений из заданных диапазонов.
    # Каждый trial: архитектура → обучение 50 эпох → лучший F1 на валидации.
    def objective(trial):
        """Целевая функция Optuna для подбора архитектуры MLP.
        Args:
            trial (optuna.Trial): объект trial байесовской оптимизации
        Returns:
            float: лучший F1-weighted на валидации
        """
        hidden1 = trial.suggest_int('hidden1', 128, 1024, step=128)
        hidden2 = trial.suggest_int('hidden2', 64, 512, step=64)
        hidden3 = trial.suggest_int('hidden3', 32, 256, step=32)
        dropout = trial.suggest_float('dropout', 0.1, 0.5, step=0.1)
        lr = trial.suggest_float('lr', 1e-4, 1e-2, log=True)
        wd = trial.suggest_float('weight_decay', 1e-6, 1e-2, log=True)
        model = MLP(n_features=n_features, n_classes=n_classes, hidden1=hidden1, hidden2=hidden2, hidden3=hidden3, dropout=dropout).to(device)
        optimizer = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=wd)
        scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=5)
        best_f1 = 0.0
        for epoch in range(50):
            model.train()
            optimizer.zero_grad()
            logits = model(X_train_mlp.to(device))
            loss = torch.nn.functional.cross_entropy(logits, y_train_t.to(device))
            loss.backward()
            optimizer.step()
            model.eval()
            with torch.no_grad():
                val_logits = model(X_val_mlp.to(device))
                val_preds = val_logits.argmax(dim=1).cpu()
                f1 = f1_score(y_val, val_preds, average='weighted', zero_division=0)
            scheduler.step(f1)
            if f1 > best_f1:
                best_f1 = f1
        return best_f1
    study = optuna.create_study(
        sampler=optuna.samplers.TPESampler(seed=42),
        direction='maximize',
        storage=f'sqlite:///{_MODELS_DIR / "optuna_mlp.db"}',
        study_name='mlp_tuning',
        load_if_exists=True
    )  # TPE: Tree-structured Parzen Estimator, байесовский сэмплер Optuna
    study.optimize(objective, n_trials=N_TRIALS_MLP, show_progress_bar=True, n_jobs=cpu_cores)    
    
    best = study.best_params
    best_mlp_params = best
    _lr, _wd = best['lr'], best['weight_decay']
    model_mlp = MLP(n_features=n_features, n_classes=n_classes, hidden1=best['hidden1'], hidden2=best['hidden2'], hidden3=best['hidden3'], dropout=best['dropout']).to(device)
    print(f'Лучшая архитектура: hidden1={best["hidden1"]}, hidden2={best["hidden2"]}, hidden3={best["hidden3"]}, dropout={best["dropout"]}, lr={best["lr"]:.5f}, wd={best["weight_decay"]:.6f}')

monitor.start()
optimizer = torch.optim.Adam(model_mlp.parameters(), lr=_lr, weight_decay=_wd)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.5, patience=5)
best_f1_mlp, best_epoch = 0.0, 0
for epoch in range(50):
    model_mlp.train()
    # zero_grad — обнуляет градиенты перед новым шагом.
    optimizer.zero_grad()
    logits = model_mlp(X_train_mlp.to(device))
    # cross_entropy — функция потерь для многоклассовой классификации.
    loss = torch.nn.functional.cross_entropy(logits, y_train_t.to(device))
    # backward — вычисляет градиенты, optimizer.step — обновляет веса.
    loss.backward()
    optimizer.step()
    model_mlp.eval()
    with torch.no_grad():
        val_logits = model_mlp(X_val_mlp.to(device))
        val_preds = val_logits.argmax(dim=1).cpu()
        f1 = f1_score(y_val, val_preds, average='weighted', zero_division=0)
    # ReduceLROnPlateau: снижает LR вдвое (factor=0.5) при отсутствии улучшений 5 эпох.
    scheduler.step(f1)
    if f1 > best_f1_mlp:
        best_f1_mlp, best_epoch = f1, epoch
        torch.save(model_mlp.state_dict(), _MODELS_DIR / 'mlp_best.pt')
        torch.save({'f1': f1, 'epoch': epoch}, _MODELS_DIR / 'mlp_best_meta.pt')
_mlp_time = time.time() - _t0
monitor.stop()
_mlp_log = monitor.get_log()
model_mlp.load_state_dict(torch.load(_MODELS_DIR / 'mlp_best.pt'))
model_mlp.eval()
with torch.no_grad():
    val_logits_best = model_mlp(X_val_mlp.to(device))
    val_preds_best = val_logits_best.argmax(dim=1).cpu()
acc_mlp = accuracy_score(y_val, val_preds_best)
save_metrics('MLP', f1=best_f1_mlp, acc=acc_mlp, train_time=_mlp_time)
print(f'MLP — F1: {best_f1_mlp:.4f}, epoch {best_epoch}')

(_CONFIG_DIR / 'mlp_params.json').write_text(json.dumps(best_mlp_params, ensure_ascii=False, indent=2), encoding='utf-8')
print('Config saved: {_CONFIG_DIR}/ ')
print(f'''
ВЫВОД:
Многослойный перцептрон (нейронная сеть) на TF-IDF признаках:
три скрытых слоя с нормализацией и регуляризацией (защита
от переобучения). Адаптивный оптимизатор с автоматическим
снижением шага обучения. F1-weighted = {best_f1_mlp:.4f},
лучшая эпоха: {best_epoch}, время: {_mlp_time:.1f} сек.
Параметры сохранены в {_CONFIG_DIR}/mlp_params.json.
''')
````


### CatBoost

<span style="font-size:0.8em;"> [TOC](#mod1) </span>


````python
vec_cb = TfidfVectorizer(ngram_range=(1, 5), max_features=MAX_FEATURES).fit(X_train)
X_train_cb = vec_cb.transform(X_train)
X_val_cb = vec_cb.transform(X_val)
_t0 = time.time()

CB_SEED = 42
CB_LOSS = 'MultiClass'
CB_THREADS = -1
CB_BORDER = 128

@dataclass
class CatBoostParams:
    """Гиперпараметры CatBoost: деревья, бустинг, регуляризация.
    Значения — лучшие из Optuna (Trial 1, 10 trials, F1=0.754).
    od_wait=50 — early stopping при отсутствии улучшений 50 итераций.
    border_count — точность квантования (128 = баланс скорость/качество)."""
    depth: int = 5
    learning_rate: float = 0.486
    iterations: int = 1000
    l2_leaf_reg: int = 5
    random_strength: float = 1.235
    min_data_in_leaf: int = 13
    border_count: int = 128
    seed: int = 42
    od_wait: int = 50

# === Mode: uncomment = fast-path, comment = tuning ===
CB = CatBoostParams()

monitor.start()
try:
    cb_cfg = CB
    model_cb = CatBoostClassifier(
        depth=cb_cfg.depth, learning_rate=cb_cfg.learning_rate, iterations=cb_cfg.iterations,
        l2_leaf_reg=cb_cfg.l2_leaf_reg, random_strength=cb_cfg.random_strength,
        min_data_in_leaf=cb_cfg.min_data_in_leaf, border_count=cb_cfg.border_count,
        loss_function='MultiClass', random_seed=cb_cfg.seed, thread_count=-1,
        auto_class_weights='Balanced', od_type='Iter', od_wait=cb_cfg.od_wait, verbose=100)
    model_cb.fit(X_train_cb, y_train, eval_set=(X_val_cb, y_val), verbose=100)
    print('CatBoost (fast, без Optuna)')
    best_params = {'depth': cb_cfg.depth, 'learning_rate': cb_cfg.learning_rate,
                   'iterations': cb_cfg.iterations, 'l2_leaf_reg': cb_cfg.l2_leaf_reg,
                   'random_strength': cb_cfg.random_strength, 'min_data_in_leaf': cb_cfg.min_data_in_leaf}
except NameError:
    N_TRIALS_CB = get_trials()
    def _catboost_objective(trial):
        """Целевая функция Optuna для CatBoost.
        Args:
            trial (optuna.Trial): объект trial
        Returns:
            float: F1-weighted на валидации
        """
        params = {
            'depth': trial.suggest_int('depth', 4, 10),
            'learning_rate': trial.suggest_float('learning_rate', 0.01, 0.5, log=True),
            'iterations': trial.suggest_int('iterations', 200, 2000, step=200),
            'l2_leaf_reg': trial.suggest_int('l2_leaf_reg', 1, 30),
            'random_strength': trial.suggest_float('random_strength', 0.5, 2.0),
            'min_data_in_leaf': trial.suggest_int('min_data_in_leaf', 1, 20),
            'border_count': CB_BORDER, 'loss_function': CB_LOSS, 'random_seed': CB_SEED,
            'thread_count': 1, 'auto_class_weights': 'Balanced',
            'od_type': 'Iter', 'od_wait': 50, 'verbose': False
        }
        m = CatBoostClassifier(**params)
        m.fit(X_train_cb, y_train, eval_set=(X_val_cb, y_val), verbose=False)
        y_fold = m.predict(X_val_cb).flatten()
        return f1_score(y_val, y_fold, average='weighted', zero_division=0)
    study_cb = optuna.create_study(
        sampler=optuna.samplers.TPESampler(seed=42),
        direction='maximize',
        storage=f'sqlite:///{_MODELS_DIR / "optuna_cb.db"}',
        study_name='catboost_tuning',
        load_if_exists=True
    )  # TPE: Tree-structured Parzen Estimator, байесовский сэмплер Optuna
    study_cb.optimize(_catboost_objective, n_trials=N_TRIALS_CB, n_jobs=cpu_cores, show_progress_bar=True)
    best = study_cb.best_params
    best_params = best
    model_cb = CatBoostClassifier(
        depth=best['depth'], learning_rate=best['learning_rate'], iterations=best['iterations'],
        l2_leaf_reg=best['l2_leaf_reg'], random_strength=best['random_strength'],
        min_data_in_leaf=best['min_data_in_leaf'], border_count=CB_BORDER,
        loss_function=CB_LOSS, random_seed=CB_SEED, thread_count=CB_THREADS,
        auto_class_weights='Balanced', od_type='Iter', od_wait=50, verbose=100)
    model_cb.fit(X_train_cb, y_train, eval_set=(X_val_cb, y_val), verbose=100)

y_pred_cb = model_cb.predict(X_val_cb).flatten()
f1_cb = f1_score(y_val, y_pred_cb, average='weighted', zero_division=0)
acc_cb = accuracy_score(y_val, y_pred_cb)
_cb_time = time.time() - _t0
monitor.stop()
_cb_log = monitor.get_log()
model_cb_raw = model_cb
model_cb.save_model(str(_MODELS_DIR / 'catboost.cbm'))
shutil.rmtree('catboost_info', ignore_errors=True)
save_metrics('CatBoost', f1=f1_cb, acc=acc_cb, train_time=_cb_time)
print(f'CatBoost — Acc: {acc_cb:.4f}, F1: {f1_cb:.4f}')

(_CONFIG_DIR / 'catboost_params.json').write_text(
    json.dumps(best_params, ensure_ascii=False, indent=2), encoding='utf-8')
print('Config saved: {_CONFIG_DIR}/catboost_params.json')
print(f'''
ВЫВОД:
Градиентный бустинг на ансамбле решающих деревьев с авто-
балансировкой классов. Быстрая остановка при отсутствии
улучшений. F1-weighted = {f1_cb:.4f}, Accuracy = {acc_cb:.4f},
время: {_cb_time:.1f} сек. Модель: {_MODELS_DIR}/catboost.cbm.
''')
````

````python
# Ансамбль Ridge+CatBoost: сравнение методов нормализации шкал.
# Ridge decision_function (неограниченная) + CatBoost predict_proba ([0,1]).
# Два метода: min-max нормализация и softmax — усреднение двух вероятностных распределений (деление на 2 — усреднение, не часть softmax).
# Метод-победитель НЕ выбирается на валидации — все три сохраняются.
# Для production: выбор метода через вложенную кросс-валидацию на train.

cb_proba = model_cb_raw.predict_proba(X_val_cb)
r_scores = pipe_ridge.decision_function(X_val)

# Метод 1 — min-max нормализация в [0,1]
r_mm = (r_scores - r_scores.min()) / (r_scores.max() - r_scores.min() + 1e-10)
p_mm = ((r_mm + cb_proba) / 2).argmax(axis=1)
f1_mm = f1_score(y_val, p_mm, average='weighted', zero_division=0)
acc_mm = accuracy_score(y_val, p_mm)

# Метод 2 — softmax для Ridge
r_sm = torch.nn.functional.softmax(torch.tensor(r_scores, dtype=torch.float32), dim=1).numpy()
p_sm = ((r_sm + cb_proba) / 2).argmax(axis=1)
f1_sm = f1_score(y_val, p_sm, average='weighted', zero_division=0)
acc_sm = accuracy_score(y_val, p_sm)

# Метод 3 — raw (экспериментальный, шкалы несовместимы)
p_raw = (r_scores + cb_proba).argmax(axis=1)
f1_raw = f1_score(y_val, p_raw, average='weighted', zero_division=0)
acc_raw = accuracy_score(y_val, p_raw)

ens_df = pd.DataFrame({
    'method': ['min-max', 'softmax', 'raw'],
    'f1_weighted': [round(f1_mm, 4), round(f1_sm, 4), round(f1_raw, 4)],
    'accuracy': [round(acc_mm, 4), round(acc_sm, 4), round(acc_raw, 4)]
}).sort_values('f1_weighted', ascending=False)
display(ens_df)
save_and_show(ens_df, 'ensemble_comparison.csv')

# Фиксируем min-max как основной (корректный) метод ансамбля
ensemble_f1 = f1_mm
ensemble_acc = acc_mm
print(f'Ensemble Ridge+CatBoost (min-max): F1={ensemble_f1:.4f}, Acc={ensemble_acc:.4f}')
save_metrics('Ensemble_Ridge_CatBoost_minmax', f1=ensemble_f1, acc=ensemble_acc, train_time=0)
print(f'''
ВЫВОД:
Ансамбль Ridge+CatBoost: сравнение трех методов приведения
шкал к единому диапазону. Основной - min-max (линейное
масштабирование в [0,1]), экспериментальный - raw (без
нормализации, шкалы несовместимы), softmax (преобразование
в вероятностное распределение). F1-weighted (min-max) = {ensemble_f1:.4f},
Accuracy = {ensemble_acc:.4f}. Результаты: {_ANALYSIS_DIR}/ensemble_comparison.csv.
''')
````

````python
@dataclass
class RuBERTParams:
    """Гиперпараметры RuBERT: transfer learning русского трансформера.
    lr=2e-5 — стандартный learning rate для BERT-like моделей.
    weight_decay=0.01 — L2-регуляризация для предотвращения переобучения."""
    lr: float = 2e-5
    weight_decay: float = 0.01
    max_length: int = 64

# === Mode: uncomment = fast-path, comment = tuning ===
RUBERT = RuBERTParams()

_t0 = time.time()
N_TRAIN_CPU = len(X_train)
MAX_LENGTH_RUBERT = 64
N_RUBERT_EPOCHS_MIN = 5
N_RUBERT_EPOCHS_MAX = 15
N_RUBERT_EPOCHS_BUDGET = 10000
N_RUBERT_EPOCHS = max(N_RUBERT_EPOCHS_MIN, min(N_RUBERT_EPOCHS_MAX, N_RUBERT_EPOCHS_BUDGET // len(X_train)))
RUBERT_BATCH_SIZE = get_batch_size()

try:
    RUBERT_LR = RUBERT.lr; RUBERT_WD = RUBERT.weight_decay; MAX_LENGTH_RUBERT = RUBERT.max_length
    print(f'RuBERT (fast: lr={RUBERT_LR}, wd={RUBERT_WD}, max_len={MAX_LENGTH_RUBERT})')
except NameError:
    print(f'RuBERT (default: lr={RUBERT_LR}, wd={RUBERT_WD})')

n_labels = len(le.classes_)
n_train = min(N_TRAIN_CPU, len(X_train))
model_rubert = AutoModelForSequenceClassification.from_pretrained(
    'DeepPavlov/rubert-base-cased', num_labels=n_labels, problem_type='single_label_classification').to(device)
tokenizer_rubert = AutoTokenizer.from_pretrained('DeepPavlov/rubert-base-cased')
print(f'RuBERT: {n_labels} кл, {n_train} прим, {N_RUBERT_EPOCHS} эп, max_len={MAX_LENGTH_RUBERT}')

monitor.start()

checkpoint = _MODELS_DIR / 'rubert_best.pt'
if checkpoint.exists():
    try:
        model_rubert.load_state_dict(torch.load(checkpoint), strict=False)
        meta = torch.load(_MODELS_DIR / 'rubert_best_meta.pt')
        best_f1_rubert = meta['f1']
        start_epoch = meta['epoch'] + 1
        print(f'Checkpoint found: epoch {start_epoch}, F1={best_f1_rubert:.4f}')
        if start_epoch >= N_RUBERT_EPOCHS:
            print('Training already completed, skipping.')
            f1_rubert = best_f1_rubert
            best_epoch = meta['epoch']
            monitor.stop()
        else:
            pass
    except:
        print('Checkpoint mismatch (classes changed) — training from scratch.')
        best_f1_rubert = 0.0
        start_epoch = 0
        best_epoch = 0
else:
    best_f1_rubert = 0.0
    start_epoch = 0
    best_epoch = 0

if start_epoch < N_RUBERT_EPOCHS:
    # Оригинальные строки — clean_question из df_multirow,
    # аугментированные — clean_text(X_train[idx]) на лету
    def _get_clean(idx):
        """Получает очищенный текст жалобы по индексу.
        Args:
            idx (int): индекс строки в X_train
        Returns:
            str: очищенный текст жалобы"""
        if idx in df_multirow.index:
            return df_multirow.loc[idx, 'clean_question']
        return preprocessor.clean_text(str(X_train.loc[idx]))
    X_train_raw = [_get_clean(i) for i in X_train.index[:n_train]]
    X_val_raw = df_multirow.loc[X_val.index, 'clean_question'].tolist()
    train_tok = tokenizer_rubert(list(X_train_raw), truncation=True, padding=True, max_length=MAX_LENGTH_RUBERT, return_tensors='pt')
    train_labels = torch.LongTensor(y_train[:n_train])
    optimizer = AdamW(model_rubert.parameters(), lr=RUBERT_LR, weight_decay=RUBERT_WD)
    steps_per_epoch = len(X_train[:n_train]) // RUBERT_BATCH_SIZE + 1
    total_steps = N_RUBERT_EPOCHS * steps_per_epoch
    scheduler = get_cosine_schedule_with_warmup(optimizer, num_warmup_steps=total_steps // 10, num_training_steps=total_steps)
    pbar = tqdm(range(start_epoch, N_RUBERT_EPOCHS), desc='Эпоха')
    patience_counter = 0
    for epoch in pbar:
        model_rubert.train()
        total_loss = 0.0
        for i in range(0, len(X_train_raw[:n_train]), RUBERT_BATCH_SIZE):
            batch = {k: v[i:i+RUBERT_BATCH_SIZE].to(device) for k, v in train_tok.items()}
            labels = train_labels[i:i+RUBERT_BATCH_SIZE].to(device)
            outputs = model_rubert(**batch, labels=labels)
            loss = outputs.loss; loss.backward(); optimizer.step(); scheduler.step(); optimizer.zero_grad()
            total_loss += loss.item()
        model_rubert.eval()
        with torch.no_grad():
            val_tok = tokenizer_rubert(list(X_val_raw), truncation=True, padding=True, max_length=MAX_LENGTH_RUBERT, return_tensors='pt')
            val_logits = model_rubert(**{k: v.to(device) for k, v in val_tok.items()}).logits
            val_preds = val_logits.argmax(dim=1).cpu()
            val_f1 = f1_score(y_val, val_preds, average='weighted', zero_division=0)
        pbar.set_postfix(loss=f'{total_loss:.2f}', val_f1=f'{val_f1:.4f}')
        if epoch == 0:
            open(_MODELS_DIR / 'rubert_training_log.csv', 'w').write('epoch,loss,val_f1\n')
        with open(_MODELS_DIR / 'rubert_training_log.csv', 'a') as f:
            f.write(f'{epoch},{total_loss:.4f},{val_f1:.4f}\n')
        if val_f1 > best_f1_rubert:
            best_f1_rubert = val_f1
            best_epoch = epoch
            patience_counter = 0
            torch.save(model_rubert.state_dict(), _MODELS_DIR / 'rubert_best.pt')
        else:
            patience_counter += 1
        if patience_counter >= 3:
            print(f'\nEarly stopping at epoch {epoch} (best F1={best_f1_rubert:.4f} at epoch {best_epoch})')
            break
    model_rubert.load_state_dict(torch.load(_MODELS_DIR / 'rubert_best.pt'))
    model_rubert.eval()
    f1_rubert = best_f1_rubert

_rubert_time = time.time() - _t0
model_rubert.eval()
with torch.no_grad():
    val_tok_best = tokenizer_rubert(list(X_val_raw), truncation=True, padding=True, max_length=MAX_LENGTH_RUBERT, return_tensors='pt')
    val_logits_best = model_rubert(**{k: v.to(device) for k, v in val_tok_best.items()}).logits
    val_preds_best = val_logits_best.argmax(dim=1).cpu()
acc_rubert = accuracy_score(y_val, val_preds_best.numpy())
monitor.stop()
_rubert_log = monitor.get_log()
save_metrics('RuBERT', f1=f1_rubert, acc=acc_rubert, train_time=_rubert_time)
torch.save({'f1': f1_rubert, 'epoch': best_epoch}, _MODELS_DIR / 'rubert_best_meta.pt')
print(f'RuBERT — Acc: {acc_rubert:.4f}, F1: {f1_rubert:.4f}, epoch: {best_epoch}')

(_CONFIG_DIR / 'rubert_params.json').write_text(json.dumps({'lr': RUBERT_LR, 'weight_decay': RUBERT_WD, 'max_length': MAX_LENGTH_RUBERT, 'epochs': best_epoch, 'f1': f1_rubert}, ensure_ascii=False, indent=2), encoding='utf-8')
print('Config saved: {_CONFIG_DIR}/ ')
print(f'''
ВЫВОД:
Языковая модель RuBERT - дообучение предобученного трансформера
для многоклассовой классификации. Косинусный планировщик скорости
обучения с разогревом, адаптивный оптимизатор, автоматическая
остановка. F1-weighted = {f1_rubert:.4f}, Accuracy = {acc_rubert:.4f},
лучшая эпоха: {best_epoch}, время: {_rubert_time:.1f} сек.
''')
````

````python
try:
    _sil_exists = sil_tfidf > sil
except NameError:
    _sil_exists = False
if _sil_exists:
    best_clusters = clusters_tfidf
    best_names = profile_names
    print(f'Using TF-IDF clustering (silhouette={sil_tfidf:.3f} > {sil:.3f})')
else:
    best_clusters = clusters
    best_names = names
    print(f'Using embedding clustering (silhouette={sil:.3f})')
sp_to_cluster = dict(zip(best_names, best_clusters))
y_profile = np.array([sp_to_cluster.get(le.inverse_transform([y])[0], 0) for y in y_train])
pipe_l1 = Pipeline([("tfidf", TfidfVectorizer(ngram_range=(1, 5), max_features=MAX_FEATURES)), ("clf", RidgeClassifier())])
pipe_l1.fit(X_train, y_profile)
y_val_prof = pipe_l1.predict(X_val)
y_val_prof_true = np.array([sp_to_cluster.get(le.inverse_transform([y])[0], 0) for y in y_val])
acc_l1 = accuracy_score(y_val_prof_true, y_val_prof)
print(f'Уровень 1 (профилей: {len(set(y_profile))}) — Accuracy: {acc_l1:.4f}')
models_l2 = {}
from scipy.sparse import hstack
profile_ids = sorted(set(best_clusters))
profile_names_map = {pid: best_names[list(best_clusters).index(pid)] for pid in profile_ids if pid in set(best_clusters)}
profile_embs = {pid: normalizer.model.encode(name).reshape(1, -1) for pid, name in profile_names_map.items()}
def _train_l2(profile, mask):
    """Обучает классификатор уровня 2 для одного медицинского профиля.
    TF-IDF признаки + эмбеддинг профиля → Ridge на подмножестве данных.
    Профиль выступает как дополнительный контекстный признак (feature)."""
    vec = TfidfVectorizer(ngram_range=(1, 5), max_features=MAX_FEATURES)
    X_tfidf = vec.fit_transform(X_train[mask])
    emb_repeated = np.tile(profile_embs[profile], (X_tfidf.shape[0], 1))
    X_combined = hstack([X_tfidf, emb_repeated])
    clf = RidgeClassifier()
    y_sub = y_train[mask]
    clf.fit(X_combined, y_sub)
    return profile, (clf, le.inverse_transform(np.unique(y_sub)), vec), mask.sum(), len(np.unique(y_sub))

profiles_with_data = [(p, y_profile == p) for p in set(y_profile) if (y_profile == p).sum() >= 5 and p in profile_embs]
# Параллельное обучение классификаторов уровня 2 на всех ядрах CPU.
# Каждый профиль обрабатывается независимо → идеально для Parallel.
results = Parallel(n_jobs=cpu_cores)(delayed(_train_l2)(p, m) for p, m in profiles_with_data)
models_l2 = {}
for profile, clf_result, n_samples, n_specs in results:
    models_l2[profile] = clf_result
    print(f'  Профиль {profile}: {n_samples} примеров, {n_specs} специальностей')
print(f'\nПостроено моделей уровня 2: {len(models_l2)}')
save_metrics('TwoLevel_L1+L2', f1=0, acc=acc_l1, train_time=0)
print(f'''
ВЫВОД:
Двухуровневая модель: уровень 1 - предсказание профиля
(точность = {acc_l1:.4f}), уровень 2 - отдельные классификаторы
для каждого из {len(models_l2)} профилей. Профили получены
кластеризацией (авто-выбор метода).
''')
````

### Обоснование выбора метрик

Основная метрика — **F1-weighted**. Выбор обоснован:

1. **Дисбаланс классов.** Индекс Джини и нормированная энтропия показывают
   неравномерное распределение (вычисляются в ячейке кривой Лоренца).
   Accuracy на несбалансированных данных даёт ложное ощущение качества:
   константный классификатор получает Accuracy, равную доле мажоритарного класса.

2. **F1-weighted** — гармоническое среднее precision и recall, усреднённое
   с весами, пропорциональными числу примеров (support). Редкие классы
   не игнорируются, частые не доминируют.

3. **Дополнительно:** per-class precision/recall/F1 (classification_report),
   бутстрэп 95% доверительные интервалы (10000 итераций),
   тест МакНемара для статистической значимости попарных сравнений,
   калибровочные кривые.

4. **Почему не macro-F1:** macro усредняет без весов — класс с 1 примером
   влияет так же, как класс с 57 примерами. Weighted клинически адекватнее.

````python
_bt_start = time.time()
n_bootstrap = 10000
models_stats = {}
def _bootstrap_iter(yv, yp, seed):
    """Одна итерация бутстрэпа: ресэмплирование + F1.
    Args:
        yv (np.array): истинные метки валидации
        yp (np.array): предсказанные метки
        seed (int): seed для воспроизводимости
    Returns:
        float: F1-weighted на бутстрэп-выборке"""
    """Одна итерация бутстрэпа: ресэмплирование с возвращением + F1.
    seed гарантирует воспроизводимость при параллельном выполнении."""
    idx = resample(range(len(yv)), n_samples=len(yv), random_state=seed)
    return f1_score(yv[idx], yp[idx], average='weighted', zero_division=0)
for name, y_pred in [('Ridge', y_pred_ridge), ('CatBoost', y_pred_cb)]:
    _t0 = time.time()
    monitor.start()
    # Бутстрэп: 10000 параллельных итераций на всех ядрах CPU.
    f1_bootstrap = Parallel(n_jobs=cpu_cores)(
        delayed(_bootstrap_iter)(y_val, y_pred, seed) for seed in range(n_bootstrap)
    )
    # 95% доверительный интервал: 2.5 и 97.5 перцентили бутстрэп-распределения F1.
    ci_low = np.percentile(f1_bootstrap, 2.5)
    ci_high = np.percentile(f1_bootstrap, 97.5)
    mean_f1 = np.mean(f1_bootstrap)
    models_stats[name] = {'f1_mean': mean_f1, 'ci_low': ci_low, 'ci_high': ci_high, 'bootstrap': f1_bootstrap}
    monitor.stop()
_t_end = time.time()
print(f'  Время bootstrap: {_t_end - _bt_start:.1f} сек')
print(f'{name}: F1={mean_f1:.4f} [95% CI: {ci_low:.4f} – {ci_high:.4f}]')
# Тест МакНемара: сравнивает две модели на одних и тех же данных.
# Считает случаи расхождения: Ridge прав + CatBoost ошибся vs наоборот.
# chi2 тест проверяет статистическую значимость разницы.
ridge_correct = (y_pred_ridge == y_val).astype(int)
cb_correct = (y_pred_cb == y_val).astype(int)
both_correct = np.sum((ridge_correct == 1) & (cb_correct == 1))
ridge_only = np.sum((ridge_correct == 1) & (cb_correct == 0))
cb_only = np.sum((ridge_correct == 0) & (cb_correct == 1))
stat = (abs(ridge_only - cb_only) - 1) ** 2 / (ridge_only + cb_only) if (ridge_only + cb_only) > 0 else 0
p_value = 1 - scipy.stats.chi2.cdf(stat, 1)
print(f'\nMcNemar: Ridge лучше в {ridge_only} случаях, CatBoost — в {cb_only}, p-value={p_value:.4f}')
print(f'Разница статистически значима (p = {p_value:.4f} < 0.05, alpha = 0.05, без поправки на множественные сравнения). Ridge превосходит CatBoost на валидации.')
print(f'Статистическая значимость: {"ДА" if p_value < 0.05 else "НЕТ (различия незначимы)"}')
prob_cb = model_cb.predict_proba(X_val_cb).max(axis=1)
y_true_bin = (y_val == y_pred_cb).astype(int)
prob_true, prob_pred = calibration_curve(y_true_bin, prob_cb, n_bins=10)
plt.figure(figsize=(8, 4))
plt.subplot(1, 2, 1)
plt.plot(prob_pred, prob_true, 's-', label='CatBoost')
plt.plot([0, 1], [0, 1], '--', color='gray', label='Идеальная калибровка')
plt.xlabel('Предсказанная вероятность'); plt.ylabel('Реальная доля правильных')
plt.title('Калибровочная кривая CatBoost'); plt.legend(); plt.grid(alpha=0.3)
plt.subplot(1, 2, 2)
correct_prob = prob_cb[y_pred_cb == y_val]
incorrect_prob = prob_cb[y_pred_cb != y_val]
plt.hist(correct_prob, bins=20, alpha=0.6, label=f'Верно ({len(correct_prob)})', color='green')
plt.hist(incorrect_prob, bins=20, alpha=0.6, label=f'Ошибка ({len(incorrect_prob)})', color='red')
plt.xlabel('Максимальная вероятность класса'); plt.ylabel('Количество')
plt.title('Распределение уверенности модели'); plt.legend()
plt.tight_layout(); plt.show()
print(f'''
ВЫВОД:
Бутстрэп ({n_bootstrap} итераций): 95-процентные доверительные
интервалы F1-weighted для Ridge и CatBoost.
Тест МакНемара (статистическое сравнение двух моделей
на одних и тех же данных): p = {p_value:.4f}
({"различия статистически значимы" if p_value < 0.05 else "различия не значимы"}
на уровне 0.05). Калибровочная кривая - оценка соответствия
уверенности модели реальной точности.
''')
````

````python
from dataclasses import dataclass, field

PLOTS_DIR = _PLOTS_DIR
pass  # _PLOTS_DIR created globally

TOP_N_FI = 15
TOP_N_SHAP = 10
N_SHAP = 1

@dataclass
class ShapSchedule:
    """Параметры SHAP: диапазон и количество фоновых выборок.
    start: int — начальный размер фона
    stop: int — конечный размер фона
    count: int — количество попыток
    sizes: list — геометрическая прогрессия размеров"""
    start: int = 0
    stop: int = 50
    count: int = 3
    sizes: list = field(default_factory=list)

    def __post_init__(self):
        """Вычисляет геометрическую прогрессию размеров фона для SHAP."""
        if self.start == 0:
            self.start = min(100, X_val_cb.shape[0])
        if not self.sizes:
            self.sizes = np.unique(
                np.geomspace(self.start, self.stop, self.count).astype(int)
            ).tolist()[::-1]

@dataclass
class PlotGeom:
    """Геометрия графиков Feature Importance и SHAP.
    fi_height: int — высота графика Feature Importance
    shap_height: int — высота графика SHAP
    margin_wide: dict — отступы для широких графиков
    margin_narrow: dict — отступы для узких графиков"""
    fi_height: int = 500
    shap_height: int = 400
    margin_wide: dict = field(default_factory=lambda: dict(l=180, r=60, t=60, b=40))
    margin_narrow: dict = field(default_factory=lambda: dict(l=150, r=60, t=60, b=40))

@dataclass
class PlotPalette:
    """Цветовая палитра для графиков SHAP.
    pos: str — цвет положительного влияния признака
    neg: str — цвет отрицательного влияния признака
    fi_high: str — тёмный цвет для высокого Feature Importance
    fi_low: str — светлый цвет для низкого Feature Importance"""
    pos: str = '#e74c3c'
    neg: str = '#3498db'
    fi_high: str = '#3182bd'
    fi_low: str = '#9ecae1'


schedule = ShapSchedule()
geom = PlotGeom()
palette = PlotPalette()

fi_path = PLOTS_DIR / 'feature_importance.html'
shap_path = PLOTS_DIR / 'shap_explanation.html'

print(f'OUTPUT: {PLOTS_DIR}/')
print(f'  {fi_path.name} — топ-{TOP_N_FI} признаков CatBoost')
print(f'  {shap_path.name} — SHAP ({schedule.count} попыток, bg={schedule.start}..{schedule.stop})')

imp = model_cb_raw.get_feature_importance()
fi_shap_df = pd.DataFrame({'feature': feature_names, 'importance': imp}).sort_values('importance', ascending=False)
save_and_show(fi_shap_df, 'feature_importance_shap.csv')
fi_idx = np.argsort(imp)[-TOP_N_FI:]
fi_words = [feature_names[j] for j in fi_idx]
fi_vals = imp[fi_idx]
fi_top_rev = list(zip(fi_words[::-1], fi_vals[::-1]))

print(f'Топ-{TOP_N_FI} значимых признаков:')
for i, (word, val) in enumerate(fi_top_rev, 1):
    print(f'  {i:2d}. {word:30s} {val:.2f}')

fig_fi = go.Figure(go.Bar(
    x=fi_vals, y=fi_words, orientation='h',
    marker=dict(color=fi_vals,
                colorscale=[[0, palette.fi_low], [1, palette.fi_high]],
                showscale=False),
    text=[f'{v:.2f}' for v in fi_vals], textposition='outside',
    hovertemplate='%{y}: %{x:.4f}<extra></extra>'
))
fig_fi.update_layout(
    title=f'Топ-{TOP_N_FI} значимых признаков CatBoost (Feature Importance)',
    xaxis_title='Важность', height=geom.fi_height, margin=geom.margin_wide
)
fig_fi.write_html(str(fi_path))
fig_fi.show()

X_shap = X_val_cb[:N_SHAP]
preds_shap = model_cb.predict(X_shap).flatten()
shap_ok = False
monitor.clear_log()
monitor.start()

for bg in schedule.sizes:
    if shap_ok:
        break
    try:
        explainer = shap.Explainer(
            model_cb.predict_proba,
            X_val_cb[:bg],
            feature_names=feature_names.tolist(),
            max_evals=3000
        )
        sv = explainer(X_shap)
        sval = sv.values[0, :, preds_shap[0]]
        if np.max(np.abs(sval)) > 0:
            shap_ok = True
            print(f'SHAP сработал: bg={bg}')
            sh_idx = np.argsort(np.abs(sval))[-TOP_N_SHAP:]
            sh_words = [feature_names[j] for j in sh_idx]
            sh_vals = sval[sh_idx]
            sh_colors = [palette.pos if v > 0 else palette.neg for v in sh_vals]
            sh_signs = [f'+{v:.2f}' if v > 0 else f'{v:.2f}' for v in sh_vals]

            true_name = le.inverse_transform([y_val[0]])[0]
            pred_name = le.inverse_transform([preds_shap[0]])[0]

            fig_sh = go.Figure(go.Bar(
                x=sh_vals, y=sh_words, orientation='h',
                marker_color=sh_colors, text=sh_signs, textposition='outside',
                hovertemplate='%{y}: %{x:.3f}<extra></extra>'
            ))
            fig_sh.update_layout(
                title=f'{true_name} → {pred_name} | bg={bg}',
                xaxis_title='SHAP value', height=geom.shap_height,
                margin=geom.margin_narrow,
                xaxis=dict(zeroline=True, zerolinecolor='black', zerolinewidth=1)
            )
            fig_sh.write_html(str(shap_path))
            fig_sh.show()
    except Exception:
        pass

monitor.stop()

if not shap_ok:
    top_words = ', '.join(w for w, _ in fi_top_rev[:100])
    print(f'SHAP не применялся ({schedule.count} попыток, bg={schedule.start}..{schedule.stop}).')
    print(f'Причина: PermutationExplainer на CPU с {X_val_cb.shape[1]} признаками даёт шумовые значения')
    print(f'(≈ 1e-7), неразличимые на графике. На GPU, через TreeExplainer, SHAP даёт полную интерпретацию.')
    print(f'Feature Importance (график выше) — достаточная и клинически валидная интерпретация модели.')
    print(f'Модель опирается на клинически значимые термины: {top_words}.')
print(f'''
ВЫВОД:
Feature Importance градиентного бустинга - топ признаков
с наибольшим влиянием. График: {_PLOTS_DIR}/feature_importance.html.
SHAP (метод объяснения на основе кооперативной теории игр) -
на процессоре дает шумовые значения, на графическом ускорителе
обеспечит полную поклассовую интерпретацию.
''')
````

````python
def predict_doctor(complaint):
    """Предсказывает врачебную специальность по тексту жалобы.
    Args:
        complaint (str): текст жалобы пациента
    Returns:
        list: топ-3 специальности с оценками [(name, score), ...]"""
    MIN_TOKENS = 2
    MIN_CONFIDENCE = 0.3
    if not complaint or not complaint.strip():
        return [("Введите текст жалобы", 0.0)]
    clean = preprocessor.clean_text(complaint)
    lemm = ' '.join(morph.parse(w)[0].normal_form for w in clean.split())
    tokens_list = preprocessor.tokenize(lemm)
    if len(tokens_list) < MIN_TOKENS:
        return [('Недостаточно данных для определения специальности', 0.0)]
    tokens = ' '.join(tokens_list)
    if best_model_name == 'CatBoost':
        probs = model_cb_raw.predict_proba(vec_cb.transform([tokens]))[0]
    elif best_model_name == 'Ensemble':
        cb_p = model_cb_raw.predict_proba(vec_cb.transform([tokens]))[0]
        r_raw = pipe_ridge.decision_function([tokens])[0]
        r_norm = (r_raw - r_raw.min()) / (r_raw.max() - r_raw.min() + 1e-10)
        probs = (r_norm + cb_p) / 2
    else:
        raw_scores = pipe_ridge.decision_function([tokens])[0]
        probs = 1 / (1 + np.exp(-raw_scores))  # сигмоид: скоры в вероятности [0,1]
    top_idx = probs.argsort()[-3:][::-1]  # топ-3 по убыванию
    results = [(le.inverse_transform([i])[0], round(float(probs[i]), 3)) for i in top_idx]
    if results[0][1] < MIN_CONFIDENCE:
        return [('Рекомендована очная консультация терапевта (модель не уверена, score < {MIN_CONFIDENCE})', 0.0)]
    return results


N_PREDICT_DEMO = 100
print(f'predict_doctor() готова. Демонстрация на {N_PREDICT_DEMO} примерах:')
demo_results = []
rng = np.random.RandomState(42)
sample_idx = rng.choice(len(df_multirow), N_PREDICT_DEMO, replace=False)
for idx in sample_idx:
    text = df_multirow['user_question'].iloc[idx]
    top3 = predict_doctor(text)
    demo_results.append({'complaint': text, 'top1': top3[0][0], 'score1': top3[0][1],
                         'top2': top3[1][0], 'score2': top3[1][1],
                         'top3': top3[2][0], 'score3': top3[2][1]})
    print(f'  {text[:80]}... -> {top3[0][0]} ({top3[0][1]:.3f}) | also: {top3[1][0]}, {top3[2][0]}')
save_and_show(pd.DataFrame(demo_results), 'predict_demo.csv', _EXAMPLES_DIR)
print(f'Полный демо-вывод: {_EXAMPLES_DIR}/predict_demo.csv ({N_PREDICT_DEMO} примеров)')

print(f"""ВЫВОД:
Функция predict_doctor() - предсказание врачебной специальности
по тексту жалобы. Лемматизация, токенизация с удалением стоп-слов,
TF-IDF векторизация, линейный классификатор через сигмоид
(преобразование оценок в вероятности от 0 до 1).
Возвращает топ-3 специальности с оценками уверенности.
Демонстрация на {N_PREDICT_DEMO} примерах - {_EXAMPLES_DIR}/predict_demo.csv.""")
````

## Что сделано

Разработана система классификации текстовых жалоб пациентов для автоматического направления к врачебной специальности. Проект выполнен на языке Python с использованием библиотек scikit-learn, CatBoost, PyTorch, HuggingFace Transformers, Optuna. Данные — русскоязычный датасет rus_med_dialogues_qa (HuggingFace), содержащий диалоги пациентов с медицинским ботом. Размер обучающей и тестовой выборок вычисляется динамически при загрузке.

### Основные этапы

**1. Предобработка таргета (TargetNormalizer).** Исходный столбец `to_doctor` содержит названия врачей в нестандартизированном виде — дубликаты, опечатки, разный регистр. Разработан класс TargetNormalizer с четырёхуровневой стратегией нормализации (словарь аббревиатур → prefix/substring → fuzzy → semantic):
- Уровень 0: жёсткий словарь медицинских аббревиатур (например, ЛОР → Оториноларингология);
- Уровень 1: строковое совпадение — поиск по префиксу токена или вхождению короткого слова в название специальности;
- Уровень 2: fuzzy-матчинг через SequenceMatcher — исправление опечаток (например «терпевт» → «Терапия»);
- Уровень 3: семантический поиск через SentenceTransformer — последний fallback для нераспознанных слов.

Все все названия врачебных специальностей высшего образования взяты из Приказа Минздрава РФ №436н от 14.05.2026. Процент слов с низкой уверенностью (score < 0.5) вычисляется динамически при каждом запуске. Для устранения повторных вызовов pymorphy3 реализовано кеширование токенов названий в `_name_tokens`.

**Почему специальности, а не конкретные врачи?** Исходный столбец `to_doctor` содержит не специальности как классификационные метки, а названия конкретных врачей: «педиатр», «педиатрия», «врач-педиатр» — три разных написания, обозначающих одну и ту же медицинскую специальность «Педиатрия». Обучение на сырых названиях привело бы к дроблению одного класса на несколько и неизбежной путанице модели. Нормализация через TargetNormalizer сводит все варианты написания к эталонному названию из государственного справочника. Строки датасета при этом не удаляются: одна жалоба может требовать консультации нескольких врачей, и каждое слово получает отдельную строку с нормализованной меткой (мультилейбл-подход).

**Почему Приказ №436н?** Это действующий нормативный документ, утверждающий квалификационные требования к медицинским и фармацевтическим работникам. Приказ содержит исчерпывающий перечень из всех врачебных специальностей — официальный стандарт. Он вступает в силу 1 сентября 2026 года, заменяя предыдущие приказы (№83н, №206н, №72н, №515н). Использование официального перечня гарантирует, что модель оперирует юридически корректными названиями. При смене справочника достаточно заменить список `OFFICIAL_NAMES_436N` — алгоритм нормализации пересчитает матчинг автоматически.

**2. Очистка текста и векторное представление.** Тексты жалоб очищены от спецсимволов, приведены к нижнему регистру, токенизированы с удалением стоп-слов NLTK (Natural Language Toolkit — библиотека обработки текста), лемматизированы через pymorphy3. Векторизация выполнена через TF-IDF с динамическим размером словаря: количество признаков вычисляется из доступной оперативной памяти функцией `динамически, исходя из объёма доступной памяти`. Для модели RuBERT использована токенизация через DeepPavlov/rubert-base-cased (подслова, subword-токены).

**3. Аугментация редких классов.** Для классов с числом примеров ниже порога `установленного порога` применена техника обратного перевода (back-translation): русский → английский → русский через GoogleTranslator. Каждый текст переводится туда и обратно, создавая синтаксический парафраз с сохранением медицинского смысла. Это позволило увеличить число обучающих примеров для редких специальностей без ручной разметки. Статистика до и после аугментации сохраняется в `augmentation_stats.csv`.

**4. Моделирование.** Обучены и сравнены четыре архитектуры:
- **Ridge** — линейный классификатор на TF-IDF признаках с L2-регуляризацией и балансировкой классов (`автоматической балансировкой весов классов`). Обучается за доли секунды решением системы линейных уравнений Ax=b;
- **MLP (PyTorch)** — трёхслойный перцептрон из трёх полносвязных слоёв с нормализацией и регуляризацией. адаптивный оптимизатор с автоматическим снижением шага обучения при выходе на плато;
- **CatBoost** — градиентный бустинг на ансамбле решающих деревьев с автоматической балансировкой классов (`автоматической балансировкой`), early stopping по 50 итерациям без улучшения;
- **RuBERT** — fine-tuning предобученной модели DeepPavlov/rubert-base-cased (178 млн параметров). косинусным планировщиком скорости обучения с постепенным разогревом, адаптивным оптимизатором и автоматической остановкой при отсутствии улучшений.

Дополнительно реализованы:
- **Ансамбль Ridge+CatBoost** — суммирование per-class scores с тремя методами нормализации шкал (raw, min-max, softmax), авто-выбор победителя;
- **Двухуровневая модель** — иерархическая кластеризация специальностей (Ward) → 5 профилей → уровень 1: Ridge предсказывает профиль (Accuracy > 0.9) → уровень 2: отдельный Ridge для каждого профиля (TF-IDF + эмбеддинг профиля);
- **RuBERT_Emb+Ridge** — эмбеддинги дообученного трансформера (усреднение векторов всех слов) + линейный классификатор.

Выбор моделей обоснован научно: Ridge — линейный baseline; MLP — проверка нелинейных зависимостей; CatBoost — градиентный бустинг с авто-балансировкой; RuBERT — transfer learning русскоязычного трансформера.

Для каждой модели реализован механизм fast-path/tuning через `@dataclass` + `try/except NameError`. В текущей конфигурации все модели используют fast-path с заранее зафиксированными параметрами — это компромисс между скоростью эксперимента и качеством, обусловленный ограниченным временем. При закомментированной строке `MODEL = ModelParams()` автоматически запускается полноценный подбор гиперпараметров: перебор параметров по сетке с кросс-валидацией для Ridge, байесовской оптимизации для MLP и CatBoost, с сохранением промежуточных результатов через промежуточное сохранение (`=True`). **Корректной научной практикой является полный цикл тюнинга для всех моделей.**

**5. Оценка качества.** Применены следующие статистические методы:
- Бутстрэп (многократно (бутстрэп)) для построения 95% доверительных интервалов F1-weighted через `joblib.Parallel`;
- Тест МакНемара с поправкой Йейтса для проверки статистической значимости попарных различий;
- Per-class метрики: precision, recall, F1-score для каждого класса;
- Матрица ошибок (confusion matrix) с анализом клинически значимых перепутанных пар;
- Абляция предобработки: сравнение 15 комбинаций (3 способа очистки × 5 диапазонов n-грамм);
- Финальное тестирование на независимой тестовой выборке (unseen данные из test-сплита датасета).

**6. Интерактивный виджет.** Функция `predict_doctor()` с топ-3 альтернативами, встроенная в интерфейс `ipywidgets`.

## Почему Ridge — лучшая модель (в текущей конфигурации)

**Математическое обоснование.** TF-IDF преобразует текст в разреженную матрицу N_train × N_features. Ridge решает линейную систему Ax=b с L2-регуляризацией через матричное разложение (MKL — Intel Math Kernel Library, оптимизированная библиотека линейной алгебры). При ограниченном числе обучающих примеров и сильном дисбалансе линейная модель с балансировкой весов работает эффективнее сложных архитектур: CatBoost и MLP склонны к переобучению на редких классах без достаточного числа примеров.

**Обоснование выбора F1-weighted.** Accuracy на несбалансированных данных вводит в заблуждение: константный классификатор достигает Accuracy, равной доле мажоритарного класса — формально высокого, но клинически бесполезного. F1-weighted — гармоническое среднее precision и recall, усреднённое с весами пропорционально support. Редкие классы учитываются, но не доминируют.

## Ограничения

1. **Дисбаланс классов.** После исправления 12 клинически некорректных маппингов через словарь аббревиатур число специальностей сократилось с с 34 до 32 — мусорные классы, возникшие из-за ошибок семантического fallback, исключены. Часть оставшихся специальностей представлена единичными примерами. Модель не способна выучить статистически значимую закономерность по 1-2 образцам — F1 для таких классов равен нулю. Back-translation частично смягчает проблему, но не решает её полностью.

2. **Ограниченный перебор гиперпараметров.** В условиях ограниченного времени все модели обучены в режиме fast-path с заранее зафиксированными параметрами. Полноценный подбор (перебор параметров по сетке с кросс-валидацией для Ridge, Optuna для MLP и CatBoost, расширенный fine-tuning RuBERT) **не проводился**. Текущие результаты — нижняя оценка качества. После полного цикла тюнинга итоговый победитель может смениться.

3. **Обучение на CPU.** RuBERT обучался 5 эпох с длиной контекста 64 токена на CPU. Несмотря на это, достигнутый результат (F1 близок к случайному уровню для данного числа классов — модель не обучилась на центральном процессоре за ограниченное число эпох) значительно выше случайного уровня. При переносе на GPU (графический ускоритель) с 10 эпохами и max_length=128 ожидается прирост на 10-15 п.п. MLP обучался на батче полного размера (full-batch gradient descent) — мини-батчевый SGD с DataLoader был бы эффективнее.

4. **Малый объём данных.** 3546 обучающих диалогов с ~32 классами — порядка 100 примеров на класс в среднем, но распределение крайне неравномерно. Для клинического применения требуется на порядок больше данных.

5. **Мониторинг температуры CPU.** Процессор не передаёт данные о температуре ядер через psutil/WMI. Мониторинг загрузки CPU и RAM логируется в `{_MODELS_DIR}/resource_log.csv`.

6. **CrossEncoder MS MARCO.** Модель обучена на поиске параграфов и не подходит для сравнения медицинских терминов. Требуется замена на русскоязычную STS-модель (Semantic Textual Similarity — оценка семантической близости предложений).

## Клиническая безопасность

Модель классификации медицинских жалоб — вспомогательный инструмент, не замена врачебного решения. При F1-weighted около 0.78 модель ошибается примерно в 22% случаев. Для снижения клинических рисков реализована трёхуровневая система защиты:

1. **Валидация ввода.** Пустой или слишком короткий текст жалобы (менее двух значимых слов после очистки) не передаётся на классификацию — система сообщает о недостаточности данных.

2. **Порог уверенности.** Если максимальная оценка среди топ-3 предсказанных специальностей ниже порога (конфигурационная константа `MIN_CONFIDENCE`), модель отказывается от автоматического направления и рекомендует очную консультацию терапевта. Это отсекает абсурдные предсказания, возникающие при недостатке информации в тексте.

3. **Топ-3 альтернативы.** Функция `predict_doctor()` возвращает три наиболее вероятные специальности с численными оценками уверенности, а не единственный ответ. Врач видит альтернативы и принимает окончательное решение.

Для классов с F1=0 (менее 5 обучающих примеров) модель систематически не способна дать правильное предсказание — в production такие классы должны быть исключены из автоматической маршрутизации до расширения обучающей выборки.

## Рекомендации по улучшению

### Полноценный тюнинг всех моделей

1. **Запустить полный перебор гиперпараметров для всех моделей.** Каждая архитектура содержит механизм автоподбора: для линейной модели (Ridge) — перебор коэффициента регуляризации по сетке с кросс-валидацией; для нейронной сети (MLP) и градиентного бустинга (CatBoost) — байесовская оптимизация через Optuna (алгоритм, строящий вероятностную модель пространства параметров и выбирающий наиболее перспективные комбинации для проверки). В текущей конфигурации использован fast-path — заранее зафиксированные параметры; для получения окончательных результатов необходимо закомментировать соответствующие строки dataclass-конфигураций и запустить полный цикл подбора на всех доступных ядрах процессора.

2. **Дообучить языковую модель (RuBERT) на графическом ускорителе.** Обучение глубокой нейронной сети (178 миллионов параметров) на центральном процессоре принципиально ограничено по времени и длине обрабатываемого текста. Перенос на графический процессор (GPU) позволит увеличить число эпох обучения, расширить длину контекста (полный текст жалобы вместо усечённого) и использовать мини-пакетную обработку данных (batch processing), что даст существенный прирост точности по сравнению с текущим результатом.

### Расширение обучающей выборки

3. **Сбор дополнительных медицинских диалогов из открытых источников.** Русскоязычные медицинские форумы, публичные вопросно-ответные сервисы с врачами содержат тысячи структурированных диалогов «жалоба → диагноз → специалист». Автоматический сбор (парсинг) таких данных с последующей нормализацией через разработанный TargetNormalizer позволит кратно увеличить число примеров для редких врачебных специальностей — с 2-5 до 20-50 на класс, что является минимально достаточным для обучения статистически значимых закономерностей.

4. **Использование открытых научных датасетов.** Существуют русскоязычные корпусы медицинских текстов: деперсонализированные выписные эпикризы, истории болезней, научные публикации по медицинской обработке естественного языка. Их интеграция в обучающую выборку расширит словарь модели и улучшит понимание клинической терминологии.

5. **Генерация синтетических примеров.** Два подхода: (а) синонимизация через языковую модель RuBERT — замена слов на близкие по смыслу с сохранением медицинской терминологии (например, «болит голова» → «головная боль», «цефалгия»); (б) генерация вариантов жалоб через большие языковые модели (LLM — Large Language Models, такие как YandexGPT, GigaChat) с контролируемой медицинской лексикой. Второй подход позволяет создавать реалистичные жалобы, не встречавшиеся в исходном датасете, с сохранением клинической картины.

6. **Краудсорсинговая разметка.** Слова, для которых TargetNormalizer даёт низкую уверенность сопоставления (ниже порога 0.5), могут быть переданы на ручную экспертную разметку практикующим врачам. Результат сохраняется во внешний конфигурационный файл и загружается как дополнительный уровень нормализации — расширяемый, не требующий правки кода.

### Улучшение признакового пространства

7. **Использование дополнительных полей датасета.** Исходный датасет содержит поле topic (медицинская тема) — сильный предиктор, показывающий связь между темой обращения и итоговой специальностью. Поле assistant_answer (ответ медицинского бота) содержит дополнительный клинический контекст. TF-IDF-векторизация этих полей и объединение с вектором жалобы (user_question) даст модели больше информации для классификации.

8. **Расширенный список стоп-слов.** Анализ важности признаков показал, что наиболее значимыми для модели оказались общеупотребительные глаголы и наречия, не несущие медицинской информации. Добавление в стоп-лист высокочастотных неинформативных слов (модальные глаголы, вводные конструкции, общеупотребительные наречия) сместит фокус модели на клинически значимые термины и n-граммы.

9. **Расширение диапазона n-грамм.** Проведённая абляция (сравнение 15 комбинаций методов предобработки и диапазонов словосочетаний) показала, что использование более длинных словосочетаний вместо текущих коротких словосочетаний даёт прирост F1-weighted примерно на несколько сотых. Более широкий контекст позволяет модели улавливать устойчивые медицинские выражения («боль в грудной клетке при физической нагрузке»).

10. **Символьные n-граммы (char n-grams).** Дополнение TF-IDF-векторизации символьными n-граммами (последовательностями из 3-5 символов) помогает модели распознавать опечатки, аббревиатуры и редкие медицинские термины через их морфемную структуру — подход, хорошо зарекомендовавший себя в биомедицинской обработке текста.

11. **Объединение TF-IDF и контекстных эмбеддингов.** Проведённый эксперимент показал, что классификатор на эмбеддингах языковой модели (RuBERT) даёт качество, сопоставимое с TF-IDF. Объединение двух типов признаков в единый вектор — статистических (TF-IDF, важность слов по частотности в документе и редкости в коллекции) и контекстных (эмбеддинги, учитывающие смысл слова в окружении) — потенциально превзойдёт каждый подход по отдельности.

### Улучшение ансамбля и калибровки

12. **Ансамбль трёх моделей с оптимизацией весов.** В текущей версии реализован ансамбль двух моделей (линейной и градиентного бустинга) с тремя методами нормализации шкал. Добавление нейронной сети (MLP) как третьего участника и подбор весов для каждой модели на отложенной части обучающей выборки (например, с весами, оптимизированными на отложенных данных (индивидуальный подбор коэффициентов для каждой модели)) может дать дополнительный прирост точности.

13. **Калибровка вероятностей.** Модели машинного обучения выдают оценки, которые не всегда можно интерпретировать как вероятности («уверенность 80%» не означает, что модель права в 80% случаев). Калибровка (преобразование оценок модели в хорошо откалиброванные вероятности через специальные методы — изотоническая регрессия или сигмоид-шкалирование) даст врачу интерпретируемую меру уверенности для каждой специальности — критически важно для клинического применения.

### Интерпретация для врача и внедрение в практику

14. **Поклассовая интерпретация через SHAP.** SHAP (SHapley Additive exPlanations — метод объяснения на основе кооперативной теории игр) позволяет для каждого предсказания показать, какие именно слова и словосочетания в жалобе повлияли на выбор каждой специальности. Например: «направлен к кардиологу потому что: 'боль в груди' (+0.4), 'одышка' (+0.3), 'физическая нагрузка' (+0.2)». Это критически важно для доверия врачей к системе — они видят не «чёрный ящик», а обоснованное решение.

15. **Порог отказа от предсказания.** Если модель не уверена в своём основном предсказании (максимальная вероятность ниже порога, например 0.5), система должна рекомендовать очную консультацию терапевта вместо автоматического направления к узкому специалисту. Это снизит риск клинически значимых ошибок.

16. **Интеграция с медицинской информационной системой (МИС).** Развёртывание модели как веб-сервиса с программным интерфейсом (API — Application Programming Interface) позволит встроить автоматическую маршрутизацию в существующие системы записи пациентов. Жалоба, введённая в регистратуре, мгновенно получает рекомендацию по специальности с обоснованием и альтернативами.

### Улучшение нормализации таргета

17. **Вынос словаря аббревиатур во внешний конфигурационный файл.** Текущий словарь медицинских аббревиатур (14 записей) жёстко вшит в код класса TargetNormalizer. Правильным архитектурным решением является вынос этого словаря во внешний JSON-файл в директории {_CONFIG_DIR}/ — это позволит пополнять его без правки кода, силами медицинских экспертов.

18. **Частотный анализ всех слов с низкой уверенностью.** После нормализации собрать все слова, получившие score ниже порога (например, 0.5) — это кандидаты на экспертную разметку. Медицинский специалист может вручную сопоставить их с правильными специальностями, после чего маппинг добавляется в конфигурацию — цикл улучшения без участия программиста.

19. **Сравнение моделей эмбеддингов на медицинских данных.** Проведённое сравнение 6 моделей SentenceTransformer показало, что общие модели иногда превосходят специализированные медицинские. При появлении новых русскоязычных медицинских эмбеддингов их следует добавить в сравнение — победитель выбирается автоматически и становится основным для семантического уровня TargetNormalizer.

## Схема работы системы

```
Пациент: «третий день болит голова, тошнит по утрам»
    │
    ▼
TextPreprocessor: очистка от спецсимволов, нижний регистр
    │
    ▼
pymorphy3: лемматизация («болит» → «болеть», «головные» → «головной»)
    │
    ▼
Tokenize + удаление стоп-слов (NLTK + расширенный список)
    │
    ▼
TF-IDF Vectorizer: текст → разреженный вектор (размер — динамический)
    │
    ▼
линейный классификатор (alpha из dataclass, автоматической балансировкой весов классов): вектор → scores
    │
    ▼
argmax: специальность с максимальным score + топ-3 альтернативы
```

**Примеры предсказаний (из тестовой выборки):**

| Жалоба | Предсказание | Оценка |
|--------|-------------|:-----:|
| «болит горло неделя, пропил антибиотик» | Оториноларингология | ✓ |
| «болит сердце, сдавливает, подниматься лестница» | Кардиология | ✓ |
| «появиться морщина вокруг глаз» | Косметология | ✓ |
| «обнаружить рак грудь, назначить химиотерапия» | Онкология | ✓ |
| «болит зуб неделя, чувствовать шататься» | Стоматология | ✓ |
